### **1º Questionamento:** Quais foram os métodos de pagamentos mais utilizados em 2023?

Vamos pedir ao ChatGPT para nos auxiliar na construção desse visual. Primeiro, vamos tratar o dado gerando uma tabela resumida (que também é uma espécie de visual) e em seguida construir o 1º gráfico.

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Temos um DataFrame chamado "df" e queremos construir um visual que apresente quais métodos de pagamentos foram os mais utilizados.

Em "df", temos a coluna "metodo_pagamento" com os 4 tipos de pagamento utilizado na compra.

Inicialmente, queremos construir uma tabela chamada "metodos_de_pagamento" que resuma os dados dessa coluna e, na sequência, desejamos criar uma visualização que apresente a quantidade desses métodos de pagamento presentes na tabela "metodo_pagamento".

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Criar a tabela "metodos_de_pagamento"
metodos_de_pagamento = df['metodo_pagamento'].value_counts().reset_index()
metodos_de_pagamento.columns = ['Metodo de Pagamento', 'Quantidade']

# Exibir a tabela
metodos_de_pagamento
````

````python
# Visualização
plt.figure(figsize=(10, 6))
plt.bar(metodos_de_pagamento['Metodo de Pagamento'], metodos_de_pagamento['Quantidade'], color='skyblue')
plt.title('Quantidade de Métodos de Pagamento Utilizados')
plt.xlabel('Método de Pagamento')
plt.ylabel('Quantidade')
plt.xticks(rotation=45, ha='right')

# Exibir o gráfico
plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

Nós conseguimos observar que mesmo sem especificar o tipo de gráfico, pelo modo que apresentamos os dados ao ChatGPT ele conseguiu representar com sucesso um visual que traga as informações desejadas. Um gráfico de colunas realmente é o ideal para representação das quantidades de uma ou mais categorias de dados.

Vamos ajustar aos poucos os visuais para que fiquem adequados para o nosso público e que tenha a identidade visual da empresa.

## **2.2 - Comparando dados**

---



Agora que já criamos nosso primeiro visual, podemos prosseguir com a construção dos próximos visuais. Aqui, vamos focar em gráficos de comparação para responder a mais dois questionamentos.

### **2º Questionamento:** Qual foi o faturamento por categoria em 2023?

**Observação:** Para o faturamento, vamos considerar aqui o produto entre o preço unitário e a quantidade somadas ao frete de cada produto.

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de barras horizontais que apresente o faturamento por categoria da Zoop.

Em "df", temos as colunas:
* categoria: categorias dos produtos
* preco_unitario: preço de uma unidade do produto
* quantidade: quantidade vendida do produto
* frete: frete a ser pago no total da compra.

Desejamos criar uma visualização que apresente o faturamento de cada categoria, em que o faturamento é dado pelo produto entre o preço unitário e a quantidade somadas ao frete.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Calcular o faturamento para cada categoria
df['faturamento'] = (df['preco_unitario'] * df['quantidade']) + df['frete']

# Agrupar por categoria e calcular o faturamento total
faturamento_por_categoria = df.groupby('categoria')['faturamento'].sum().reset_index()

# Ordenar por faturamento
faturamento_por_categoria = faturamento_por_categoria.sort_values(by='faturamento', ascending=False)

# Visualização
plt.figure(figsize=(10, 8))
plt.barh(faturamento_por_categoria['categoria'], faturamento_por_categoria['faturamento'], color='skyblue')
plt.title('Faturamento por Categoria')
plt.xlabel('Faturamento')
plt.ylabel('Categoria')

# Exibir o gráfico
plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

Aqui especificamos o tipo de gráfico, guiando o ChatGPT para criação do nosso visual. Podemos notar que em um gráfico de barras o foco é maior nas categorias, além de ser uma forma mais fluida de ler textos que possam ser extensos demais para um gráfico de coluna.

### **3º Questionamento:** Como foram as vendas ao longo dos meses em 2023?

**Observação:** É importante especificar como a sua variável tempo está especificada dentro da base para o ChatGPT alcançar resultados melhores na sugestão do visual

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de linha que apresente as vendas totais mensais da Zoop.

Em "df", temos as colunas:
* data: com as datas das compras no formato datetime (`aaaa-mm-dd`)
* faturamento: dado pelo produto entre o preço unitário e a quantidade somadas ao frete de cada venda.

Desejamos criar uma visualização que apresente as vendas por mês. Primeiro, agrupe os dados por mês e depois crie uma nova coluna chamada "mes" que receba o nome de cada mês, traduzindo os meses do índice por meio do dicionário abaixo. Utilize a coluna "data" para o agrupamento dos dados e construção do gráfico. A coluna "mes" só deve ser utilizada para alterar os rótulos do eixo x.

```
meses = {
    'January': 'Jan',
    'February': 'Fev',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov',
    'December': 'Dez'
}
```

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Dicionário de tradução de meses
meses = {
    'January': 'Jan',
    'February': 'Fev',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov',
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] = df['data'].dt.strftime('%B')
vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6))
plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color='#1890FF', linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais da Zoop', fontsize=18, color='#555555')
plt.xlabel('Mês', fontsize=12, color='#1890FF')
plt.ylabel('Vendas', fontsize=12, color='#1890FF')

# Alterar rótulos do eixo x utilizando o dicionário de tradução
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=10, color='#1890FF')

# Exibir o gráfico
plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

Quando possuimos uma quantidade de períodos maior para os nossos dados, vale a pena recorrer a gráficos de linha(s) que é ideal para séries temporais. Como nossos dados são mensais, utilizar um gráfico de colunas, por exemplo, reduziria a percepção de uma sequência dos dados. Linhas dão uma maior sensação de continuidade.

Sinta-se livre para elaborar algumas hipóteses dos momentos de alta e baixa das compras.

## **2.3 - Compondo dados dinâmicos**

---



Construimos alguns visuais interessantes para comparação dos dados, trazendo os faturamentos da Zoop. Agora, vamos partir para um nova categoria de visuais: composição.

Vamos uma composição dinâmica dos dados. Para isso, vamos adicionar mais variáveis à análise, trazendo 3 diferentes variáveis sendo uma delas categórica.

### **4º Questionamento:** Como esta dividida a venda por trimestre por método de pagamento? Existe um período em que algum tipo de pagamento se sobressai?

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de barras empilhadas que apresente as vendas por trimestre em relação ao método de pagamento da Zoop.

Em "df", temos as colunas:
* data: com as datas das compras no formato datetime (`aaaa-mm-dd`)
* faturamento: com os valores das vendas
* metodo_pagamento: com o método de pagamento escolhido na compra

Desejamos criar uma visualização que apresente as vendas por trimestre, agrupando os métodos de pagamento de cada trimestre correspondente a cada compra.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Converter a coluna 'data' para o formato datetime, se ainda não estiver
df['data'] = pd.to_datetime(df['data'])

# Criar uma nova coluna para identificar o trimestre
df['trimestre'] = df['data'].dt.to_period('Q')

# Agrupar por trimestre e método de pagamento, somando o faturamento
vendas_por_trimestre = df.groupby(['trimestre', 'metodo_pagamento'])['faturamento'].sum().unstack()

# Visualização em gráfico de barras empilhadas
vendas_por_trimestre.plot(kind='bar', stacked=True, figsize=(12, 8), colormap='viridis')
plt.title('Vendas por Trimestre e Método de Pagamento')
plt.xlabel('Trimestre')
plt.ylabel('Vendas')
plt.legend(title='Método de Pagamento')

plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

O gráfico de colunas empilhadas adiciona à comparação entre períodos ou categorias a comparação também entre os dados de cada período/categoria. É como se colocássemos uma "lupa" em cada coluna e vizualizássemos a composição de cada uma delas de acordo com a variável desejada (aqui, as formas de pagamento).

Sinta-se livre para elaborar algumas hipóteses das vendas pelo trimestre combinados ao tipo de pagamento e quais insights podemos retirar dessa base.

Vamos inicialmente trazer um exemplo de como podemos utilizar a comparação de dados estáticos, ou seja, um recorte do tempo ou das variáveis que possuímos.

### **5º Questionamento:** Qual a proporção da clientela que participa de nosso programa de cashback?

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Temos um DataFrame chamado "df" e queremos construir um visual que apresente a proporção dos(as) usuários(as) que estão aderindo o programa de cashback da Zoop.

Em "df", temos a coluna "cashback" que dispõe de tais dados.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Calcular a proporção de adesão ao cashback
proporcao_cashback = df['cashback'].value_counts(normalize=True)

# Visualização
plt.figure(figsize=(6, 6))
plt.pie(proporcao_cashback, labels=proporcao_cashback.index, autopct='%1.1f%%', colors=['skyblue', 'lightcoral'])
plt.title('Proporção de Usuários Aderindo ao Cashback')
plt.show()
````

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Você conseguiria inverter as cores e alterar o gráfico anterior para um do tipo rosca?

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Calcular a proporção de adesão ao cashback
proporcao_cashback = df['cashback'].value_counts(normalize=True)

# Cores invertidas
cores_invertidas = ['lightcoral', 'skyblue']

# Visualização em gráfico de rosca (donut)
plt.figure(figsize=(6, 6))
plt.pie(proporcao_cashback, labels=proporcao_cashback.index, autopct='%1.1f%%', startangle=90, colors=cores_invertidas, wedgeprops=dict(width=0.6))
plt.title('Proporção de Usuários Aderindo ao Cashback')
plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

Olha só! Mais uma vez a IA nos apresentou uma boa noção de qual gráfico aplicar para esse ponto, mas você conseguiu notar que "ela" não faz ideia dos dados que possuímos?

Precisamos estar atentos(as) a isso, pois passamos apenas os nomes das colunas e que dados existiam lá. Ela não consegue saber o que é um valor de caráter positivo ou negativo, por exemplo. Por isso, o "não" estava em azul e o "sim" em vermelho.  No prompt seguinte, demos uma orientação para inverter as cores (ainda sem falar o real valor dos dados) e pedimos uma alteração do visual.

Que tal você experimentar fazer outros pedidos e ver como o ChatGPT se comporta?

Falando do gráfico, gráficos de setores são ideais para poucas categorias de dados, se possível dados dicotômicos (2 categorias, como sim e não, gato e cachorro...). Eles representam a proporção dos dados de maneira estática.

### **7º Questionamento:** Como está distribuído o nosso público por idade e sexo biológico? Existe um grupo que sobressai?

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Temos um DataFrame chamado "df" e queremos construir um visual que apresente a distribuição do nosso público por idade e sexo biológico.

Em "df", temos as colunas:
* idade: idade da pessoa que realizou a compra
* sexo_biologico: sexo biológico da pessoa que realizou a compra, com dados categóricos

Desejamos criar uma visualização que apresente a distribuição desses dados.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Supondo que você tenha o DataFrame "df" carregado

# Visualização em gráfico de barras empilhadas
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='idade', hue='sexo_biologico', multiple='stack', bins=20, palette='Set2')
plt.title('Distribuição do Público por Idade e Sexo Biológico')
plt.xlabel('Idade')
plt.ylabel('Contagem')
plt.legend(title='Sexo Biológico')

plt.show()
````

O gráfico gerado é pouco prático, pois compila os dados do sexo biológico "Masculino" e "Feminino" em uma mesma barra! Precisamos trazer mais uma instrução ao GPT para escolher um visual mais adequado.

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Acho que podemos melhorar esse gráfico. Que tal mudar para um boxplot?

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Supondo que você tenha o DataFrame "df" carregado

# Visualização em boxplot
plt.figure(figsize=(12, 8))
sns.boxplot(data=df, x='sexo_biologico', y='idade', palette='pastel')
plt.title('Distribuição da Idade por Sexo Biológico')
plt.xlabel('Sexo Biológico')
plt.ylabel('Idade')

plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

Dessa vez foi necessário guiar a IA a fim de gerar um visual melhor para nossas análises.

O gráfico aqui representado é um boxplot (diagrama de caixa) que é ideal para representar um conjunto de dados de acordo com 5 medidas de interesse: valor máximo, mínimo, 1º quartil (onde estão situados os primeiros 25% dos dados), 2º quartil ou mediana (onde estão situados 50% dos dados) e 3º quartil (onde estão situados 75% dos dados).

Lembra do `describe()`? Este gráfico apresenta em um único elemento boa parte dos dados que a função exibe em tabela.

Agora chegou a hora de refinarmos as nossas visualizações. Vamos começar com o 1º gráfico que fizemos, aquele com os métodos de pagamento?

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA_1, as colunas na cor VERMELHO_1 e o tamanho da figura deve ser de (10, 6).

O título da figura é "Métodos de pagamentos mais utilizados em 2023" com tamanho 18 e sua cor CINZA_5.

Os rótulos do eixo y (yticks) devem ser suprimidos assim como o label do eixo y (ylabel) e os rótulos e label do eixo X (xticks e xlabel) devem ter tamanho 12 com a cor AZUL_1. As bordas do topo, lado direito e esquerdo do gráfico devem ser suprimidas.

Além disso, é necessário exibir o total de cada método de pagamento acima de cada coluna, com o texto também na cor AZUL_1 e tamanho 12.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização. Lembre-se que as variáveis das cores que estamos utilizando possui valores de cores já em hexadecimal.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Agrupar por método de pagamento e calcular a contagem
metodos_pagamento = df['metodo_pagamento'].value_counts()

# Configurações do gráfico
plt.figure(figsize=(10, 6), facecolor=CINZA_1)
ax = plt.axes()
ax.set_facecolor(CINZA_1)
plt.bar(metodos_pagamento.index, metodos_pagamento, color=VERMELHO_1)

# Adicionar o total acima de cada coluna
for i, v in enumerate(metodos_pagamento):
    plt.text(i, v + 0.2, str(v), color=AZUL_1, ha='center', va='bottom', fontsize=12)

# Configurações adicionais
plt.title('Métodos de Pagamentos mais Utilizados em 2023', fontsize=18, color=CINZA_5)
plt.xticks(fontsize=12, color=AZUL_1)
plt.yticks([])
plt.ylabel('')
plt.xlabel('Método de Pagamento', fontsize=12, color=AZUL_1)
ax.spines[['top', 'right', 'left']].set_visible(False)

plt.show()
````

## **4.3 - Adicionando textos**

---



Já refinamos um dos gráficos na etapa anterior. Agora, vamos trazer esse gráfico, adicionando mais um elemento que pode ser importante às análises: textos informativos. Vamos utilizar a IA novamente para esse processo.

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

O gráfico ficou muito bom!

Agora quero adicionar um texto dentro da área do plot, vamos lá?

Para isso, some os valores das duas primeiras colunas, divida pela soma de totas as colunas e passe para o formato de porcentagem armazenando na variável "percentual"

Agora, na posição x = 2 e y = 2500, adicione o seguinte texto com tamanho 10 e cor AZUL_3: `f"$\\bf{percentual:.2f}$% dos clientes utilizam Cartão de Crédito ou PIX \npara pagamento. Seria interessante recorrer a parcerias\ncom bancos para a construção de um Zoop Pay a fim\nde fidelizar a nossa clientela."`.

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Agrupar por método de pagamento e contar o total de ocorrências
metodos_de_pagamento = df['metodo_pagamento'].value_counts()

# Calcular percentual
total_clientes = metodos_de_pagamento.sum()
percentual = (metodos_de_pagamento[0] + metodos_de_pagamento[1]) / total_clientes * 100

# Configurações do gráfico
plt.figure(figsize=(10, 6), facecolor=CINZA_1)
ax = plt.axes()
ax.set_facecolor(CINZA_1)
plt.bar(metodos_de_pagamento.index, metodos_de_pagamento, color=VERMELHO_1)

# Adicionar texto com o total acima de cada barra
for i, v in enumerate(metodos_de_pagamento):
    plt.text(i, v + 0.1, str(v), ha='center', va='bottom', color=AZUL_1, fontsize=12)

# Adicionar texto dentro da área do plot
plt.text(2, 2500, f"$\\bf{percentual:.2f}$% dos clientes utilizam Cartão de Crédito ou PIX \npara pagamento. Seria interessante recorrer a parcerias\ncom bancos para a construção de um Zoop Pay a fim\nde fidelizar a nossa clientela.",
         ha='left', va='center', color=AZUL_3, fontsize=10)

# Configurações adicionais
plt.title('Métodos de Pagamentos mais Utilizados em 2023', fontsize=18, color=CINZA_5)
plt.xticks(fontsize=12, color=AZUL_1)
plt.yticks([])
plt.xlabel('Método de Pagamento', fontsize=12, color=AZUL_1)
ax.spines[['top', 'right', 'left']].set_visible(False)
plt.grid(False)

plt.show()
````

<font color="7b71ff" size=4> Comentários: </font>

---

Aproveitamos a área útil para trazer um *insight* que tivemos ao descrever os dados dos métodos de pagamento. Isso já aumenta e muito a relevância do dado representado e fornece mais informação a quem lê.

## **4.4 - Adicionando elementos visuais**

---



Bem bacana o uso dos textos informativos, não é mesmo? E se agora nós adicionarmos outros elementos visuais a um de nossos gráficos?

Vamos escolher agora o gráfico de linha (**Vendas totais mensais da Zoop**) e aplicar mais uma técnica a ele por meio da utilização da IA .

````python
import pandas as pd
import matplotlib.pyplot as plt

# Dicionário de tradução de meses
meses = {
    'January': 'Jan',
    'February': 'Fev',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov',
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] = df['data'].dt.strftime('%B')
vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6))
plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color='#1890FF', linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais da Zoop', fontsize=18, color='#555555')
plt.xlabel('Mês', fontsize=12, color='#1890FF')
plt.ylabel('Vendas', fontsize=12, color='#1890FF')

# Alterar rótulos do eixo x utilizando o dicionário de tradução
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=10, color='#1890FF')

plt.show()
````

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Vamos explorar mais um dos gráficos que fizemos? Agora vamos utilizar aquele com as Vendas totais mensais da Zoop. A variável que passa os valores mensais é:

```
df['mes'] = df['data'].dt.strftime('%B')
vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())
```

Em que `df["data"]` correspondente as datas no formato datetime e `df['faturamento']` os valores das vendas.

E o que gerou o gráfico original foi

```
plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color='#1890FF', linewidth=2)
```

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA_1, a linha na cor AZUL_1 e o tamanho da figura deve ser de (12, 6).

O título da figura é "Vendas totais mensais em 2023" com tamanho 18 e sua cor CINZA_5.

Os rótulos do eixo y (yticks) devem estar o formato "R$ valor M" com duas casas decimais, tamanho 12 e cor AZUL_1.

Os rótulos do eixo X (xticks) devem ter tamanho 12 com a cor AZUL_1. Já os labels do eixo x e y (xlabel e ylabel) devem ser suprimidos. As bordas do topo e lado direito do gráfico devem ser suprimidas.

### **Gráfico 1 - Métodos de pagamentos mais utilizados em 2023:**

````python
def metodos_pag():
  # Agrupar por método de pagamento e contar o total de ocorrências
  metodos_de_pagamento = df['metodo_pagamento'].value_counts()

  # Calcular percentual
  total_clientes = metodos_de_pagamento.sum()
  percentual = (metodos_de_pagamento[0] + metodos_de_pagamento[1]) / total_clientes * 100

  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)
  plt.bar(metodos_de_pagamento.index, metodos_de_pagamento, color=VERMELHO_1)

  # Adicionar texto com o total acima de cada barra
  for i, v in enumerate(metodos_de_pagamento):
      plt.text(i, v + 0.1, str(v), ha='center', va='bottom', color=AZUL_1, fontsize=12)

  # Adicionar texto dentro da área do plot
  plt.text(2, 2500, f"$\\bf{percentual:.2f}$% dos clientes utilizam Cartão de Crédito ou PIX \npara pagamento. Seria interessante recorrer a parcerias\ncom bancos para a construção de um Zoop Pay a fim\nde fidelizar a nossa clientela.",
          ha='left', va='center', color=AZUL_3, fontsize=10)

  # Configurações adicionais
  plt.title('Métodos de Pagamentos mais Utilizados em 2023', fontsize=18, color=CINZA_5)
  plt.xticks(fontsize=12, color=AZUL_1)
  plt.yticks([])
  plt.xlabel('Método de Pagamento', fontsize=12, color=AZUL_1)
  ax.spines[['top', 'right', 'left']].set_visible(False)
  plt.grid(False)

  return plt
````

### **Gráfico 3 - Faturamento por categoria de produto em 2023:**

Vamos dar prosseguimento aos gráficos gerados ao longo das aulas, partindo agora do gráfico de faturamento por categoria de produto aplicando as técnicas de storytelling que aprendemos aqui no curso por meio da utilização da IA

````python
import matplotlib.pyplot as plt

# Calcular o faturamento para cada categoria
df['faturamento'] = (df['preco_unitario'] * df['quantidade']) + df['frete']

# Agrupar por categoria e calcular o faturamento total
faturamento_por_categoria = df.groupby('categoria')['faturamento'].sum().reset_index()

# Ordenar por faturamento
faturamento_por_categoria = faturamento_por_categoria.sort_values(by='faturamento', ascending=False)

# Visualização
plt.figure(figsize=(10, 8))
plt.barh(faturamento_por_categoria['categoria'], faturamento_por_categoria['faturamento'], color='skyblue')
plt.title('Faturamento por Categoria')
plt.xlabel('Faturamento')
plt.ylabel('Categoria')

plt.show()
````

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Vamos partir para mais um gráfico? Agora, vamos refinar o gráfico de faturamento por categoria. Lembre que esse gráfico foi de barras horizontais.

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA_1, as colunas na cor AQUA_1 e o tamanho da figura deve ser de (10, 6).

O título da figura é "Faturamento por categoria de produto em 2023" com tamanho 18 e sua cor CINZA_5.

Os rótulos do eixo x (xticks) devem ser suprimidos assim como os labels do eixo x e y (xlabel e ylabel) e os rótulos do eixo y (yticks) devem ter tamanho 12 com a cor AZUL_1. Todas as bordas do gráfico devem ser suprimidas.

Além disso, é necessário exibir o total do faturamento de cada categoria do lado direito de cada barra, com o texto também na cor AZUL_1 e tamanho 12.

Vamos adicionar também um texto dentro da área do plot. Para isso, divida o valor da 1ª barra pela soma de todas as barras e passe para o formato de porcentagem armazenando na variável "percentual"

Agora, na posição x = 0.4e7 e y = 5, adicione o seguinte texto com tamanho 12 e cor AZUL_3: `f"$\\bf{percentual:.2f}$% das vendas correspondem apenas as vendas de Eletrônicos.\nIsso é equivalente a quase o dobro em relação a todas as outras categorias\npresentes no portfólio da Zoop."`.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import matplotlib.pyplot as plt

def faturamento_por_categoria():
  # Agrupar por categoria e calcular o faturamento total
  faturamento_por_categoria = df.groupby('categoria')['faturamento'].sum().sort_values(ascending=False)

  # Calcular percentual
  total_faturamento = faturamento_por_categoria.sum()
  percentual = faturamento_por_categoria.iloc[0] / total_faturamento * 100

  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)
  bars = plt.barh(faturamento_por_categoria.index, faturamento_por_categoria, color=AQUA_1)

  # Adicionar rótulos e título
  plt.title('Faturamento por Categoria de Produto em 2023', fontsize=18, color=CINZA_5)
  plt.xlabel('')
  plt.ylabel('')
  plt.yticks(fontsize=12, color=AZUL_1)
  plt.xticks([])
  ax.spines[['top', 'right', 'left', 'bottom']].set_visible(False)

  # Adicionar total do faturamento acima de cada barra
  for bar in bars:
      plt.text(bar.get_width() + 1e5, bar.get_y() + bar.get_height()/2,
              f'R$ {bar.get_width()/1e6:.2f} M',
              ha='left', va='center', color=AZUL_1, fontsize=12)

  # Adicionar texto dentro da área do plot
  plt.text(0.4e7, 6, f"$\\bf{percentual:.2f}$% das vendas correspondem apenas as vendas de Eletrônicos.\nIsso é equivalente a quase o dobro em relação a todas as outras categorias\npresentes no portfólio da Zoop.",
          ha='left', va='center', color=AZUL_3, fontsize=12)

  return plt
````

````python
import matplotlib.pyplot as plt

def vendas_tri_metodo():
  # Agrupar por trimestre e método de pagamento e calcular o faturamento total
  vendas_por_trimestre_metodo = df.groupby(['trimestre', 'metodo_pagamento'])['faturamento'].sum().unstack()

  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)

  # Plotar o gráfico de barras empilhadas
  vendas_por_trimestre_metodo.plot(kind='bar', stacked=True, color=[BRANCO, CINZA_3, AQUA_1, AQUA_3], ax=ax, edgecolor=CINZA_1)

  # Adicionar rótulos e título
  plt.title('Vendas por Trimestre e Método de Pagamento em 2023', fontsize=18, color=CINZA_5)
  plt.xlabel('')
  plt.ylabel('')
  plt.xticks(rotation=0, fontsize=12, color=AZUL_1)
  plt.yticks(ax.get_yticks(), ['R$ {:.0f} M'.format(val/1e6) for val in ax.get_yticks()], fontsize=12, color=AZUL_1)
  ax.spines['top'].set_visible(False)
  ax.spines['right'].set_visible(False)
  ax.spines['left'].set_visible(False)
  ax.spines['bottom'].set_visible(False)
  ax.yaxis.grid(color='gray', linestyle='--', linewidth=0.5)
  ax.set_axisbelow(True) # adicionar grid atrás

  return plt
````

### **Gráfico 5 - Clientes participantes do programa de cashback:**

Agora chegou a hora do gráfico de setores. Ele já esta bem encaminhado, mas precisamos realizar alguns ajustes.

````python
import matplotlib.pyplot as plt

# Calcular a proporção de adesão ao cashback
proporcao_cashback = df['cashback'].value_counts(normalize=True)

# Cores invertidas
cores_invertidas = ['lightcoral', 'skyblue']

# Visualização em gráfico de rosca (donut)
plt.figure(figsize=(6, 6))
plt.pie(proporcao_cashback, labels=proporcao_cashback.index, autopct='%1.1f%%', startangle=90, colors=cores_invertidas, wedgeprops=dict(width=0.6))
plt.title('Proporção de Usuários Aderindo ao Cashback')
plt.show()
````

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Agora, vamos refinar o gráfico de rosca que produzimos para a proporção de usuários aderindo ao programa de cashback.

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA_1, a cor da fatia de valor "Não" igual a CINZA_3, a cor da fatia de valor "Sim" igual a AZUL_1  e o tamanho da figura deve ser de (6, 6).

O título da figura é "Participam do programa de cashback?" com tamanho 18 e sua cor CINZA_5.

Os rótulos das fatias precisam ter a mesma cor das fatias e tamanho 12 e as porcentagens precisam estar na cor BRANCO, também com tamanho 12.

A propriedade da largura da rosca (`wedgeprops`) precisa ser de 0.4, a posição das porcentagens dentro da fatia (`pctdistance`) precisa ser de 0.8, o ângulo de início de 180 (`startangle`) e sem explosão de fatias.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import matplotlib.pyplot as plt

def cashback():
  # Calcular a proporção de usuários aderindo ao programa de cashback
  proporcao_cashback = df['cashback'].value_counts(normalize=True)

  # Configurações do gráfico de rosca
  fig, ax = plt.subplots(figsize=(6, 6), facecolor=CINZA_1)
  ax.set_facecolor(CINZA_1)

  # Plotar o gráfico de rosca
  wedges, texts, autotexts = ax.pie(proporcao_cashback, labels=proporcao_cashback.index, autopct='%1.1f%%', startangle=180,
                                    wedgeprops=dict(width=0.4, edgecolor=CINZA_1), pctdistance=0.8, colors=[CINZA_3, AZUL_1])

  # Adicionar rótulos e título
  plt.title('Participam do Programa de Cashback?', fontsize=18, color=CINZA_5)

  # Configurar rótulos e porcentagens com as cores especificadas
  for text, autotext, color in zip(texts, autotexts, [CINZA_3, AZUL_1]):
      text.set_color(color)
      text.set_fontsize(12)
      autotext.set_color(BRANCO)
      autotext.set_fontsize(12)

  return plt
````

```
f"Média de avaliações = {df['avaliacao_compra'].mean().round(2)}\n
Avaliação mais comum = {df['avaliacao_compra'].value_counts().idxmax()}\n
Frequência da mais comum = {df['avaliacao_compra'].value_counts().max()}"
```

na cor AZUL_1. Adicione a caixa da anotação com o texto na posição xytext=(2, 1800) e coloque uma seta, na cor AZUL_1, saindo da caixa até a posição xy=(9, 2250).


Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import matplotlib.pyplot as plt
import seaborn as sns

def avaliacao_compra():
  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)

  # Plotar o histograma das avaliações
  sns.histplot(df['avaliacao_compra'], bins=11, color=VERMELHO_1, kde=False, ax=ax)

  # Adicionar rótulos e título
  plt.title('Distribuição das Avaliações das Compras em 2023', fontsize=18, color=CINZA_5)
  plt.xlabel('Avaliação', fontsize=12, color=AZUL_1)
  plt.ylabel('Frequência', fontsize=12, color=AZUL_1)
  plt.xticks(range(11), fontsize=12, color=AZUL_1)
  plt.yticks(fontsize=12, color=AZUL_1)
  ax.spines['top'].set_visible(False)
  ax.spines['right'].set_visible(False)
  ax.spines['left'].set_visible(False)
  ax.spines['bottom'].set_visible(False)
  ax.set_axisbelow(True)
  ax.yaxis.grid(color='gray', linestyle='--', linewidth=0.5)

  # Adicionar caixa de anotação
  media_avaliacao = df['avaliacao_compra'].mean().round(2)
  avaliacao_mais_comum = df['avaliacao_compra'].value_counts().idxmax()
  freq_avaliacao_mais_comum = df['avaliacao_compra'].value_counts().max()

  texto_anotacao = f"Média de avaliações = {media_avaliacao}\nAvaliação mais comum = {avaliacao_mais_comum}\nFrequência da mais comum = {freq_avaliacao_mais_comum}"

  ax.annotate(texto_anotacao, xy=(9, 2250), xytext=(2, 1800), fontsize=12, color=AZUL_1,
              bbox=dict(boxstyle="round,pad=0.3", edgecolor=CINZA_3, facecolor=BRANCO),
              arrowprops=dict(facecolor=AZUL_1, arrowstyle="->", connectionstyle="arc3,rad=0.3"))

  return plt
````

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

````python
import matplotlib.pyplot as plt
import seaborn as sns

def dist_idade_sexo():
    # Configurações do gráfico
    plt.figure(figsize=(10, 6), facecolor=CINZA_1)
    ax = plt.axes()
    ax.set_facecolor(CINZA_1)

    # Plotar o boxplot da distribuição da idade por sexo biológico
    sns.boxplot(data=df, x='sexo_biologico', y='idade', palette=[AZUL_1, AQUA_1], ax=ax)

    # Adicionar rótulos e título
    plt.title('Distribuição da Idade por Sexo Biológico dos(as) clientes', fontsize=18, color=CINZA_5)
    plt.xlabel('')
    plt.ylabel('Idade', fontsize=12, color=AZUL_1)
    plt.xticks(fontsize=12, color=AZUL_1)
    plt.yticks(fontsize=12, color=AZUL_1)
    ax.spines[['top', 'right', 'left', 'bottom']].set_visible(False)
    ax.set_axisbelow(True)
    ax.yaxis.grid(color='gray', linestyle='--', linewidth=0.5)

    # Adicionar caixas de anotação
    texto_anotacao1 = f"Mínimo = {df.groupby('sexo_biologico')['idade'].min()[1]}\n" \
                    f"1º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.25)[1]}\n" \
                    f"2º quartil = {df.groupby('sexo_biologico')['idade'].median()[1]}\n" \
                    f"3º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.75)[1]}\n" \
                    f"Máximo = {df.groupby('sexo_biologico')['idade'].max()[1]}"

    texto_anotacao2 = f"Mínimo = {df.groupby('sexo_biologico')['idade'].min()[0]}\n" \
                    f"1º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.25)[0]}\n" \
                    f"2º quartil = {df.groupby('sexo_biologico')['idade'].median()[0]}\n" \
                    f"3º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.75)[0]}\n" \
                    f"Máximo = {df.groupby('sexo_biologico')['idade'].max()[0]}"

    ax.annotate(texto_anotacao1, xy=(-0.1, 45), xytext=(-0.4, 50), fontsize=12, color=AZUL_1,
                bbox=dict(boxstyle="round,pad=0.3", edgecolor=CINZA_3, facecolor=BRANCO),
                arrowprops=dict(facecolor=AZUL_1, arrowstyle="->", connectionstyle="arc3,rad=0.3"))

    ax.annotate(texto_anotacao2, xy=(0.9, 45), xytext=(0.6, 50), fontsize=12, color=AZUL_1,
                bbox=dict(boxstyle="round,pad=0.3", edgecolor=CINZA_3, facecolor=BRANCO),
                arrowprops=dict(facecolor=AZUL_1, arrowstyle="->", connectionstyle="arc3,rad=0.3"))

    return plt
````

# Para saber mais: como criar um bom prompt no ChatGPT

Uma das habilidades mais importantes para a interação com IAs generativas como o ChatGPT é a capacidade de criar bons prompts guiando a ferramenta de acordo com o seu conhecimento sobre o negócio. A criação de um prompt eficaz no ChatGPT é fundamental para obter resultados precisos e relevantes para resolução dos problemas.

No contexto de um projeto de análise de dados, como o que estamos trabalhando aqui, um prompt bem elaborado pode ser a chave para obter insights valiosos. Aqui, estão algumas boas práticas e exemplos levando em conta um projeto simples de Data Science:

    - Seja específico e direto:

Explique diretamente o que você deseja que o ChatGPT execute. Isso ajuda a IA a entender melhor sua necessidade, fornecendo respostas mais precisas.

Exemplo: “Execute uma análise exploratória de dados na base ‘zoop_vendas.csv' usando as bibliotecas de manipulação de dados do Python. Apresente um resumo estatístico das vendas por região”

     - Forneça um contexto quando necessário

Informe ao ChatGPT o contexto do projeto, incluindo dados, objetivos e restrições. Isto pode melhorar significativamente a qualidade das respostas. No entanto, evite informações desnecessárias que possam confundir a IA.

Exemplo: "Estou trabalhando em um projeto de análise de vendas de uma loja de varejo. A base de dados possui informações de vendas diárias por categoria de produto e região em 2023. Me ajude a calcular a soma das vendas de “Eletrônicos” por trimestre."

     - Especifique o formato da resposta:

Indique o formato desejado para a resposta, como gráficos, tabelas ou explicações detalhadas.

Exemplo: "Crie um gráfico de linhas que mostra a evolução das vendas ao longo do último ano, segmentado por categoria de produtos."

     - Indique o nível de detalhe desejado

Deixe evidente o nível de detalhe ou a profundidade da explicação que você está procurando. Isso ajuda a IA a ajustar a resposta à sua necessidade de compreensão.

Exemplo: "Pode me fornecer um exemplo comentado de como usar a função groupby( ) do pandas para agrupar dados por categoria e calcular a soma total de vendas para cada grupo?"

     - Use termos técnicos corretos

Esteja atento a terminologia específica relacionada ao seu campo de interesse. Isso assegura que a IA compreenda o domínio de sua questão e responda de forma adequada.

Exemplo: “Como implemento um modelo de regressão linear no Python usando scikit-learn para prever vendas futuras com base em gastos com publicidade?”

     - Divida tarefas complexas em etapas simples:

Se o projeto envolve várias etapas, divida-o em partes menores e instrua o ChatGPT passo a passo. Isso torna mais fácil para a IA ajudar de forma eficaz.

Exemplo:

  - "Como posso lidar com valores ausentes em meu conjunto de dados usando pandas?"

  - "Qual é a melhor maneira de visualizar a distribuição de idades em meu conjunto de dados usando matplotlib?"

  - “Você poderia ajustar o visual de distribuição de idades colocando os rótulos e labels com tamanho 12 e cor #CCCCCC”

    - Esteja aberto a interações

Esteja sempre preparado(a) para fazer ajustes em seus prompts com base nas respostas recebidas. A interação pode ajudar a refinar as respostas e obter exatamente o que você precisa.

Exemplo: “Esta resposta pode ficar melhor. Que tal arredondar os valores em duas casas decimais e exibir no gráfico apenas os valores maiores que 1 milhão (1e6)?”

Essas dicas são fundamentais na construção de prompts bem instrutivos como os que você terá contato ao longo desse curso. Observe como foram descritas as etapas e ações nos prompts do curso e como você pode aplicar em seu projeto utilizando o ChatGPT.

Certifique-se de comunicar suas intenções de forma precisa, dividir tarefas complexas em etapas gerenciáveis e fornecer exemplos quando apropriado. Isso ajudará a obter respostas mais relevantes e úteis para o seu projeto, tornando a interação com o chat mais eficaz.

````python
import pandas as pd

# Supondo que voc tenha os DataFrames "clientes" e "vendas" carregados

# Unindo as bases de dados com base na coluna "ID compra"
dados_combinados = pd.merge(vendas, clientes, on-'ID_compra')

# Organizando as colunas na sequência desejada
colunas organizadas = [
    'ID_compra', 'data', 'horario', 'categoria, preco_unitario',
    'quantidade, 'frete', 'metodo_pagamento', 'ID_cliente', 'idade',
    'sexo biologico', 'cidade', 'uf, 'regiao', 'cashback', 'avaliacao_compra'
]

dados_combinados = dados_combinados[colunas_organizadas]
````

Temos algumas coisas interessantes, mas vamos fazer uma pequena alteração. Ele gerou um novo dataframe e chamou de dados_combinados. Mas podemos atribuir um nome mais fácil, como df, que é um nome mais curto e conseguimos lembrar mais facilmente para fazer os processos daqui para frente.

````python
import pandas as pd

# Supondo que voc tenha os DataFrames "clientes" e "vendas" carregados

# Unindo as bases de dados com base na coluna "ID compra"
df = pd.merge(vendas, clientes, on-'ID_compra')

# Organizando as colunas na sequência desejada
colunas organizadas = [
    'ID_compra', 'data', 'horario', 'categoria, preco_unitario',
    'quantidade, 'frete', 'metodo_pagamento', 'ID_cliente', 'idade',
    'sexo biologico', 'cidade', 'uf, 'regiao', 'cashback', 'avaliacao_compra'
]

df = df[colunas_organizadas]
````

````python

# Supondo que voc tenha os DataFrames "clientes" e "vendas" carregados

# Unindo as bases de dados com base na coluna "ID compra"
df = pd.merge(vendas, clientes, on-'ID_compra')

# Organizando as colunas na sequência desejada
colunas organizadas = [
    'ID_compra', 'data', 'horario', 'categoria, preco_unitario',
    'quantidade, 'frete', 'metodo_pagamento', 'ID_cliente', 'idade',
    'sexo biologico', 'cidade', 'uf, 'regiao', 'cashback', 'avaliacao_compra'
]

df = df[colunas_organizadas]
````

Agora, vamos unir o "df" com a base "zoop_pay", lida a partir da url "url_zoop_pay", gerando um novo "df". Para união das bases, utilizaremos a coluna "ID_cliente". Além disso, vamos ajustar a coluna "data" para o formato datetime, em que seu formato na base de dados está como "YYYY-MM-DD".

## Construindo o prompt

Começaremos com o processo de construção do gráfico da maneira mais básica possível. Para isso, responderemos o primeiro questionamento: "Quais foram os métodos de pagamento mais utilizados em 2023?".

Vamos solicitar a ajuda do Chat GPT na elaboração desse visual. No entanto, vamos dividir a criação do visual em duas partes. Primeiro, construiremos uma tabela que resuma os métodos de pagamento mais utilizados. Em seguida, criaremos o gráfico com base nessa tabela. Vale ressaltar que tanto tabelas quanto gráficos são considerados visuais. Ambos podem ser utilizados como representações simplificadas dos dados.

Ao ler o prompt, analisaremos o que está sendo solicitado e qual será a resposta que o chat GPT nos fornecerá. O prompt é:

Temos um DataFrame chamado "df" e queremos construir um visual que apresente quais métodos de pagamentos foram os mais utilizados.

Em "df", temos a coluna "metodo_pagamento" com os 4 tipos de pagamento utilizado na compra.

Inicialmente, queremos construir uma tabela chamada "metodos_de_pagamento" que resuma os dados dessa coluna e, na sequência, desejamos criar uma visualização que apresente a quantidade desses métodos de pagamento presentes na tabela "metodo_pagamento".

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

É importante deixar claro que queremos o dataframe chamado df e não desejamos nomes gerados aleatoriamente pelo chat GPT. Nosso objetivo é construir um visual que apresente quais métodos de pagamento foram os mais utilizados. No dataframe df, temos a coluna metodo_pagamento, com quatro tipos de pagamento utilizados nas compras. Além disso, especificamos a quantidade de tipos de pagamento dentro dos nossos dados. Com essas informações, o chat GPT terá uma ideia de qual visual utilizar na construção desse gráfico.

Ao analisar o prompt, percebemos que nossa primeira etapa é criar essa tabela e atribuí-la à variável metodo_pagamento. Em seguida, utilizaremos essa variável para construir nossa visualização. Vamos copiar o prompt e colá-lo no chat GPT para gerar o código responsável por criar esse visual.

Após executar o prompt no Chat GPT, obtivemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado
# Criar a tabela "metodos_de_pagamento"
metodos_de_pagamento = df['metodo_pagamento'].value_counts().reset_index()
metodos_de_pagamento.columns = ['Metodo de Pagamento', 'Quantidade']

# Visualização
plt.figure(figsize=(10, 6))
plt.bar(metodos_de_pagamento['Metodo de Pagamento'], metodos_de_pagamento['Quantidade'], color='skyblue')
plt.title('Quantidade de Métodos de Pagamento Utilizados')
plt.xlabel('Método de Pagamento')
plt.ylabel('Quantidade')
plt.xticks(rotation=45, ha='right')

# Exibir a tabela
print("Tabela 'metodos_de_pagamento':")
print(metodos_de_pagamento)

# Exibir o gráfico
plt.show()
````

É importante destacar que mencionamos as colunas que serão utilizadas e explicamos o processo para calcular o faturamento. Isso será crucial ao gerar o código, pois utilizaremos esse mesmo processo para outros casos semelhantes.

Ao executar o prompt dentro do Chat GPT, obtemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado
# Calcular o faturamento para cada categoria
df['faturamento'] = (df['preco_unitario'] * df['quantidade']) + df['frete']

# Agrupar por categoria e calcular o faturamento total
faturamento_por_categoria = df.groupby('categoria')['faturamento'].sum().reset_index()

# Ordenar por faturamento
faturamento_por_categoria = faturamento_por_categoria.sort_values(by='faturamento', ascending=False)

# Visualização
plt.figure(figsize=(12, 8))
plt.barh(faturamento_por_categoria['categoria'], faturamento_por_categoria['faturamento'], color='skyblue')
plt.title('Faturamento por Categoria')
plt.xlabel('Faturamento')
plt.ylabel('Categoria')

# Exibir o gráfico
plt.show()
````

Analisando e reorganizando o código
Incialmente, ele fez a importação das bibliotecas Pandas e Matplotlib. Como já realizamos essas importações anteriormente, vamos remover essa parte. Também vamos retirar o comentário de suposição e manter apenas o processo de criação da coluna e construção do gráfico.

Primeiramente, ele está criando dentro do nosso df uma coluna onde será armazenando o faturamento. Para isso, ele multiplica o preco_unitario pela quantidade, lendo cada uma das colunas, e soma com o valor do frete. Assim, ele gera essa nova coluna já com esses dados completos.

Em seguida, ele cria uma tabela em que calcula o faturamento por categoria. Esse faturamento por categoria é obtido através de um groupby( ) da categoria pela coluna faturamento. Ou seja, ele está agrupando todos os dados por categoria e, para agregar, faz a soma (sum) de todos os faturamentos de uma categoria passando este valor. Depois, ele repete esse processo para cada uma das categorias.

O reset_index( ) é utilizado para limpar o índice e trazer o faturamento para uma das colunas. Assim, conseguimos ter o valor tanto da categoria quanto do faturamento. Posteriormente, ele solicita a ordenação dos dados pelo faturamento, em ordem decrescente, utilizando o método sort_values( ).

Depois, ele cria uma visualização de tamanho 12 por 8 e utiliza um gráfico de barras horizontais (plt.barh). Então, mapeia o faturamento por categoria, onde a categoria é plotada no eixo y e o valor do faturamento no eixo x. Além disso, mantém as cores originais, adiciona títulos e labels.

````python
# Calcular o faturamento para cada categoria
df['faturamento'] = (df['preco_unitario'] * df['quantidade']) + df['frete']

# Agrupar por categoria e calcular o faturamento total
faturamento_por_categoria = df.groupby('categoria')['faturamento'].sum().reset_index()

# Ordenar por faturamento
faturamento_por_categoria = faturamento_por_categoria.sort_values(by='faturamento', ascending=False)

# Visualização
plt.figure(figsize=(12, 8))
plt.barh(faturamento_por_categoria['categoria'], faturamento_por_categoria['faturamento'], color='skyblue')
plt.title('Faturamento por Categoria')
plt.xlabel('Faturamento')
plt.ylabel('Categoria')

# Exibir o gráfico
plt.show()
````

Note que no eixo vertical temos os valores das diferentes categorias de produtos, enquanto no eixo horizontal estão os valores na escala de 10 milhões. Com isso, conseguimos ter uma ideia básica de que os setores de eletrônicos e eletrodomésticos representam uma grande parte do faturamento dentro das categorias.

## Construindo o prompt

Antes de construirmos o gráfico, é importante esclarecer como está estruturada a coluna de datas para o Chat GPT, para que ele possa fazer a agregação necessária para a visualização.

Vamos revisar nosso prompt e ver o que especificamos:

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de linha que apresente as vendas totais mensais da Zoop.

Em "df", temos as colunas:

  - data: com as datas das compras no formato datetime (aaaa-mm-dd)

  - faturamento: dado pelo produto entre o preço unitário e a quantidade somadas ao frete de cada venda.

Desejamos criar uma visualização que apresente as vendas por mês. Prmeiro, agrupe os dados por mês e depois crie uma nova coluna chamada "mes" que receba o nome de cada mês, traduzindo os meses do índice por meio do dicionário abaixo. Utilize a coluna "data" para o agrupamento dos dados e construção do gráfico. A coluna "mes" só deve ser utilizada para alterar os rótulos do eixo x.

meses = { 'January': 'Jan', 'February': 'Fev', 'March': 'Mar', 'April': 'Abr', 'May': 'Mai', 'June': 'Jun', 'July': 'Jul', 'August': 'Ago', 'September': 'Set', 'October': 'Out', 'November': 'Nov', 'December': 'Dez' }

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Indicamos que temos um dataframe chamado df e queremos criar uma visualização em forma de gráfico de linha, que apresente as vendas mensais da Zoop. No dataframe, temos as colunas data, contendo as datas no formato datetime (aaaa-mm-dd), e faturamento, que é a coluna que criamos no visual anterior, calculada pelo produto do preço unitário pela quantidade somada ao frete de cada venda.

Então, estamos planejando criar uma coluna mes que derivará dos dados em data, e os nomes dos meses estarão em inglês, já que DateTime os apresenta nesse idioma. Para garantir que os meses apareçam corretamente em nosso gráfico, utilizaremos um dicionário que traduzirá os nomes dos meses para o formato abreviado em português.

No entanto, faremos essa tradução depois de criar a coluna, pois se traduzirmos antes, os meses serão organizados em ordem alfabética, o que causará problemas na representação dos dados.

Preparamos um dicionário que relaciona os nomes em inglês dos meses com os correspondentes em português, também abreviados. À esquerda, temos os nomes em inglês, e à direita, os nomes em português com 3 caracteres. Além disso, solicitamos que o Chat GPT utilize as bibliotecas de Python para criar a visualização.

Vamos executar este prompt. Ao fazer isso, obtemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado
# Dicionário de tradução de meses
meses = {
    'January': 'Jan',
    'February': 'Fev',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov',
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] = df['data'].dt.strftime('%B')
vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6))
plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color='#1890FF', linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais da Zoop', fontsize=18, color='#555555')
plt.xlabel('Mês', fontsize=12, color='#1890FF')
plt.ylabel('Vendas', fontsize=12, color='#1890FF')

# Alterar rótulos do eixo x utilizando o dicionário de tradução
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=10, color='#1890FF')

# Exibir o gráfico
plt.show()
````

## Analisando e reorganizando o código

Após a geração do código, vamos remover as linhas de importação das bibliotecas e o comentário de suposição, mantendo a partir da parte em que é definido o dicionário para a tradução dos meses.

Após o dicionário, o código realiza o agrupamento dos dados. Primeiramente, é criada uma coluna mes a partir da coluna data, utilizando o método strftime( ) para converter as datas no formato de string e representar apenas o nome do mês, como November, December, etc.

Em seguida, é criada a tabela vendas_mensais, na qual os dados são agrupados pelo mes, somando o faturamento correspondente. Para garantir a ordenação correta dos meses, é utilizado o método reindex( ) com as chaves do dicionário meses.keys ( ). Isso assegura que os meses sejam ordenados de January a December, em inglês.

Posteriormente, é criada a figura com dimensões de 12 por 6, onde os valores dos meses são utilizados como índice e os valores de faturamento como os dados a serem plotados. Um marcador é adicionado para cada ponto, com cor e tamanho ajustados conforme necessário. Os rótulos e títulos do gráfico também são ajustados, embora possam ser refinados posteriormente.

Por fim, ocorre o ajuste dos rótulos do eixo x, onde os nomes dos meses em inglês são substituídos pelos equivalentes em português utilizando uma compreensão de lista. Isso é feito acessando o dicionário meses e substituindo cada mês pelo seu equivalente em português, garantindo que os meses sejam exibidos corretamente no gráfico.

````python
meses = {
    'January': 'Jan',
    'February': 'Fev',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov',
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] = df['data'].dt.strftime('%B')
vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6))
plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color='#1890FF', linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais da Zoop', fontsize=18, color='#555555')
plt.xlabel('Mês', fontsize=12, color='#1890FF')
plt.ylabel('Vendas', fontsize=12, color='#1890FF')

# Alterar rótulos do eixo x utilizando o dicionário de tradução
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=10, color='#1890FF')

# Exibir o gráfico
plt.show()
````

image.png

Optamos por utilizar um gráfico de linha porque ele proporciona uma sensação de continuidade dos dados ao longo do tempo. Com 12 meses para representar, o gráfico de linha permite visualizar a tendência de forma mais clara, enquanto um gráfico de barras poderia dificultar essa percepção devido à separação em colunas.

## Construindo o prompt

O quarto questionamento que enfrentaremos é o seguinte: "Como está distribuída a venda por trimestre de acordo com o método de pagamento? Há algum período em que um determinado método de pagamento se destaca?".

Aqui, temos três elementos distintos: as vendas, representadas pelo faturamento; os trimestres, que dividem o tempo; e os métodos de pagamento. Portanto, temos três variáveis a serem consideradas, e veremos se o ChatGPT será capaz de responder de forma eficaz à criação desse visual.

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de barras empilhadas que apresente as vendas por trimestre em relação ao método de pagamento da Zoop.

Em "df", temos as colunas:

  - data: com as datas das compras no formato datetime (aaaa-mm-dd)

  - faturamento: com os valores das vendas

  - metodo_pagamento: com o método de pagamento escolhido na compra

Desejamos criar uma visualização que apresente as vendas por trimestre, agrupando os métodos de pagamento de cada trimestre correspondente a cada compra.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Especificamos que desejamos barras empilhadas, o que significa que cada barra representará um trimestre, e dentro de cada barra, teremos sub-barras representando os diferentes métodos de pagamento da Zoop. Nosso objetivo é criar uma visualização que apresente as vendas por trimestre, agrupando os métodos de pagamento correspondentes a cada trimestre de compra.

Ao executar o prompt dentro do Chat GPT, obtemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado
# Converter a coluna 'data' para o formato datetime, se ainda não estiver
df['data'] = pd.to_datetime(df['data'])

# Criar uma nova coluna para identificar o trimestre
df['trimestre'] = df['data'].dt.to_period('Q')

# Agrupar por trimestre e método de pagamento, somando o faturamento
vendas_por_trimestre = df.groupby(['trimestre', 'metodo_pagamento'])['faturamento'].sum().unstack()

# Visualização em gráfico de barras empilhadas
vendas_por_trimestre.plot(kind='bar', stacked=True, figsize=(12, 8), colormap='viridis')

plt.title('Vendas por Trimestre e Método de Pagamento')
plt.xlabel('Trimestre')
plt.ylabel('Vendas')
plt.legend(title='Método de Pagamento')

plt.show()
````

## Analisando e reorganizando o código

O código gerado inclui a importação das bibliotecas Pandas e Matplotlib, que podemos remover, pois já foram feita anteriormente. Além disso, há uma instrução para converter a coluna data para o tipo datetime, mas isso já foi feito durante a etapa de exploração de dados, então também podemos removê-la. Portanto, vamos nos concentrar apenas na parte relevante para a construção do gráfico.

O código está criando uma nova coluna chamada trimestre, onde lê os dados da coluna data e os converte para o formato de trimestre utilizando o método to_period('Q'). Isso significa que estamos alterando a granularidade dos dados, passando de uma escala diária para uma escala trimestral, o que nos fornece uma visão mais ampla e menos detalhada dos dados.

Após isso, o código cria uma tabela chamada vendas_por_trimestre. Nessa tabela, ocorre um agrupamento por duas variáveis: trimestre e metodo_pagamento. Portanto, teremos um objeto com múltiplos índices, representando os trimestres e os métodos de pagamento. O objetivo é calcular o faturamento para cada combinação de trimestre e método de pagamento. Isso é feito através de uma agregação, onde o valor de faturamento é somado para cada grupo de trimestre e método de pagamento. Em seguida, é aplicado o método unstack( ), que remove um nível dos índices, resultando em uma tabela cruzada.

Uma tabela cruzada é uma estrutura de dados que cruza informações entre duas variáveis, organizando os valores de acordo com essas variáveis. No caso, os trimestres são colocados nas linhas e os métodos de pagamento nas colunas. Os valores dentro da tabela representam a combinação entre trimestre e método de pagamento. Esta tabela cruzada será a base para a construção do nosso gráfico.

O código utiliza o método plot( ) sobre o objeto vendas_por_trimestre, gerando um gráfico de barras. Ao passar o parâmetro stacked=True, indicamos que queremos as barras empilhadas umas sobre as outras, representando a distribuição dos dados.

O tamanho da figura é definido como 12 por 8 usando o parâmetro figsize, e um esquema de cores é escolhido aleatoriamente com o parâmetro colormap. Por fim, são adicionados títulos e rótulos ao gráfico para facilitar a compreensão.

````python
# Criar uma nova coluna para identificar o trimestre
df['trimestre'] = df['data'].dt.to_period('Q')

# Agrupar por trimestre e método de pagamento, somando o faturamento
vendas_por_trimestre = df.groupby(['trimestre', 'metodo_pagamento'])['faturamento'].sum().unstack()

# Visualização em gráfico de barras empilhadas
vendas_por_trimestre.plot(kind='bar', stacked=True, figsize=(12, 8), colormap='viridis')

plt.title('Vendas por Trimestre e Método de Pagamento')
plt.xlabel('Trimestre')
plt.ylabel('Vendas')
plt.legend(title='Método de Pagamento')

plt.show()
````

## Analisando o gráfico

Ao executar o código, obtemos o seguinte gráfico:

image.png

No eixo Y, temos os valores de vendas em milhões, variando de 1 a 8. As barras maiores representam os trimestres, indo do primeiro ao quarto trimestre de 2023. Já as barras empilhadas representam os diferentes métodos de pagamento, como boleto, cartão de crédito, cartão de débito e PIX. Essas barras mostram como os valores são compostos, fornecendo uma visão detalhada da distribuição das vendas ao longo dos trimestres e por método de pagamento.

Podemos observar, dentro da barra correspondente ao quarto trimestre, que o total de vendas quase ultrapassa 8 milhões. Dentro desse valor, se destacam os métodos de pagamento de cartão de crédito e PIX. Isso nos permite entender como os dados estão distribuídos e como cada tipo de pagamento contribui para o total das vendas.

Esse visual é bastante interessante e nos desafia a investigar por que alguns métodos de pagamento se destacam mais em determinados períodos do que em outros. Podemos criar hipóteses a partir desses dados para entender melhor esse comportamento.

# Faça como eu fiz: gráfico de colunas agrupadas

Aprendemos no vídeo anterior a construir um gráfico de colunas empilhadas para composição de dados por meio do suporte do ChatGPT. Que tal aprendermos também um outro tipo de visual que alia a composição e comparação de dados por meio de um problema?

A diretoria da Zoop está preocupada com as categorias com menor faturamento anual do seu e-commerce e, por isso, gostaria de entender a performance dessas categorias isoladamente. Para isso, nós recebemos a demanda de explorar o faturamento por trimestre buscando investigar como está o equilíbrio dessas vendas por cada categoria.

Para resolver este problema, nós vamos agrupar as 3 categorias com os menores faturamentos anualmente, separando-as por trimestre por meio de um gráfico de colunas agrupadas.

Mas, por que um gráfico de colunas agrupadas?

Um gráfico de colunas agrupadas é ideal para comparar categorias diferentes em relação a uma variável (podendo ser temporal ou não), mostrando como valores se comparam dentro de cada categoria em colunas separadas, facilitando a visualização de padrões e diferenças entre grupos.

Seguindo, vamos pedir auxílio ao ChatGPT para construir esse visual por meio do seguinte prompt:

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de colunas agrupadas lado a lado que apresente as vendas por trimestre em relação às categorias dos produtos.

Em "df", temos as colunas:

  - trimestre: com as datas em trimestres das compras

  - faturamento: com os valores das vendas

  - categoria: com as categorias dos produtos vendidos. Vamos utilizar apenas as categorias "Papelaria e escritório", "Livros" e "Alimentos e bebidas".

Desejamos criar uma visualização que apresente as vendas por trimestre ("faturamento"), agrupando as categorias lado a lado em relação ao trimestre correspondente a cada compra.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

O código gerado pelo prompt foi o seguinte:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você já tenha o DataFrame df

# Filtrar as categorias desejadas
categorias_desejadas = ["Papelaria e escritório", "Livros", "Alimentos e bebidas"]
df_filtrado = df[df['categoria'].isin(categorias_desejadas)]

# Pivotar os dados para ter os trimestres como índice e as categorias como colunas
pivot_df = df_filtrado.pivot_table(index='trimestre', columns='categoria', values='faturamento', aggfunc='sum')

# Criar o gráfico de colunas agrupadas
pivot_df.plot(kind='bar', stacked=False, figsize=(10, 6))

# Adicionar rótulos e título
plt.xlabel('Trimestre')
plt.ylabel('Vendas')
plt.title('Vendas por Trimestre e Categoria')
plt.xticks(rotation=45)  # Rotaciona os rótulos do eixo x para melhorar a legibilidade

# Mostrar o gráfico
plt.legend(title='Categoria')
plt.tight_layout()
plt.show()
````

image.png

Note que observamos em um mesmo visual como as categorias estão compostas em cada trimestre e conseguimos comparar seus resultados dentro desse período. É possível notar que tivemos um recuo nas vendas das 3 categorias no 3º tri em relação ao trimestre anterior e que grande parte do faturamento está concentrado no 4º trimestre. Podemos utilizar o seguinte prompt para a resolução do 1º desafio:

Temos um DataFrame chamado "df" e queremos construir um visual de gráfico de colunas que apresente o faturamento por região da Zoop.

Em "df", temos as colunas:

  - regiao: regiao do(a) cliente

  - preco_unitario: preço de uma unidade do produto

  - quantidade: quantidade vendida do produto

  - frete: frete a ser pago no total da compra.

Desejamos criar uma visualização que apresente o faturamento de cada região, em que o faturamento é dado pelo produto entre o preço unitário e a quantidade somadas ao frete.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Utilizando a resposta do ChatGPT, obtemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que df seja o seu DataFrame
# Calcula o faturamento para cada transação
df['faturamento'] = (df['preco_unitario'] * df['quantidade']) + df['frete']

# Agrupa os dados por região e soma o faturamento
faturamento_por_regiao = df.groupby('regiao')['faturamento'].sum().reset_index()

# Ordena o resultado para melhor visualização
faturamento_por_regiao = faturamento_por_regiao.sort_values(by='faturamento', ascending=False)

# Criação do gráfico de colunas
plt.figure(figsize=(10, 6))  # Ajusta o tamanho do gráfico
plt.bar(faturamento_por_regiao['regiao'], faturamento_por_regiao['faturamento'], color='skyblue')  # Plota o gráfico de barras
plt.title('Faturamento por Região')  # Adiciona título ao gráfico
plt.xlabel('Região')  # Adiciona rótulo ao eixo X
plt.ylabel('Faturamento')  # Adiciona rótulo ao eixo Y
plt.show()  # Exibe o gráfico
````

Agora, vamos para o 2º desafio, em que precisamos gerar um visual que demonstre o público por sexo biológico nos estados de São Paulo e Rio de Janeiro. Logo, podemos utilizar o seguinte prompt para a resolução destes processos:

Temos um DataFrame chamado "df" e queremos construir um visual de colunas empilhadas que apresente a quantidade de pessoas por sexo biológico em relação aos estados de "São Paulo" e "Rio de Janeiro".

Em "df", temos as colunas:

  - uf: unidade federativa (estado) do(a) cliente. Vamos utilizar apenas os estados de "São Paulo" e "Rio de Janeiro"

  - sexo_biologico: sexo biológico do(a) cliente
  
Desejamos criar uma visualização que apresente a quantidade de pessoas agrupadas por sexo biológico nos estados que definimos.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Utilizando a resposta do ChatGPT, obtemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Filtrar o DataFrame para incluir apenas os estados de "São Paulo" e "Rio de Janeiro"
df_filtrado = df[df['uf'].isin(['São Paulo', 'Rio de Janeiro'])]

# Contar a quantidade de pessoas por sexo biológico e estado
contagem_por_sexo_e_estado = df_filtrado.groupby(['uf', 'sexo_biologico']).size().unstack(fill_value=0)

# Criar o gráfico de colunas empilhadas
contagem_por_sexo_e_estado.plot(kind='bar', stacked=True, figsize=(10, 6))

# Adicionar título e rótulos aos eixos
plt.title('Quantidade de Pessoas por Sexo Biológico em São Paulo e Rio de Janeiro')
plt.xlabel('Estado')
plt.ylabel('Quantidade de Pessoas')
plt.xticks(rotation=0)  # Mantém os rótulos dos estados na horizontal para melhor legibilidade

# Mostrar a legenda e o gráfico
plt.legend(title='Sexo Biológico')
plt.show()
````

# Para saber mais: histogramas e gráficos de densidade

No vídeo anterior, criamos um prompt que gerou o código de um gráfico de distribuição de acordo com as orientações que passamos sobre os nossos dados. Mesmo sem termos especificado o tipo de visual, a IA definiu que o gráfico a ser desenhado seria um histograma de colunas.

Foi possível notar que o código gerado pelo ChatGPT gerou um visual bem adequado para apresentar a distribuição da amostra de dados que possuímos, em que cada coluna representava a frequência de repetição de um valor. Mas por que ele (ChatGPT) não escolheu, por exemplo, um gráfico de densidade? Como saber qual utilizar?

Vamos utilizar o próprio ChatGPT para responder estas dúvidas.

Prompt: Estou trabalhando com a geração de visuais utilizando a Linguagem Python e notei que um dos gráficos gerados foi um histograma de coluna. Porém, sei que também existe uma forma de criar visual como gráfico de densidade.

Explique porque o ChatGPT escolheu apresentar os dados de avaliações do público (valores inteiros de 0 até 10) como um histograma de coluna e não um gráfico de densidade, qual a diferença entre esses dois visuais e quando podemos escolher um ou outro. Traga exemplos simples da utilização de um histograma e gráfico de densidade, em que o tema seja de um e-commerce.

A resposta dada pelo ChatGPT é a seguinte:

A escolha entre um histograma de coluna (também conhecido como histograma de barras) e um gráfico de densidade geralmente depende do tipo de dado sendo analisado e do objetivo da visualização.

## Por que um histograma de coluna?

O histograma de coluna é amplamente utilizado para visualizar a distribuição de frequências ou contagens em um conjunto de dados. Este tipo de gráfico é particularmente eficaz para mostrar a distribuição de variáveis discretas ou categorias. No caso de avaliações do público, que são valores inteiros de 0 até 10, um histograma de coluna permite visualizar claramente quantas avaliações foram recebidas para cada ponto de avaliação específico. Isso facilita a identificação de tendências, como a concentração de avaliações em determinadas notas, e ajuda a compreender a distribuição geral das avaliações.

## Diferença entre Histograma de Coluna e Gráfico de Densidade

  - Histograma de Coluna: Utiliza barras verticais para representar a frequência de cada categoria ou intervalo de valores. A altura de cada barra indica a contagem ou proporção de observações naquela categoria.

  - Gráfico de Densidade: Utiliza uma linha para traçar a distribuição contínua dos dados. Este tipo de gráfico é mais adequado para variáveis contínuas, onde mostra a probabilidade de valores dentro de um continuum, fornecendo uma ideia da densidade dos dados.

## Quando escolher cada um?

  - Histograma de Coluna: É ideal para dados categóricos ou discretos, quando se deseja visualizar a frequência ou contagem por categoria. É útil quando as categorias são poucas e claramente definidas, como notas de 0 a 10.

  - Gráfico de Densidade: É mais adequado para variáveis contínuas, quando se deseja entender a distribuição e densidade dos dados ao longo de um intervalo. É útil para visualizar a forma da distribuição dos dados, como normal, binomial, etc.

## Exemplos de Utilização em E-commerce

  - Histograma de Coluna: Imagine que você deseja analisar as avaliações de clientes para um produto específico em seu e-commerce. As avaliações vão de 0 a 5. Um histograma de coluna pode mostrar quantos clientes deram cada nota, permitindo identificar rapidamente a satisfação geral dos clientes com o produto.

  - Gráfico de Densidade: Se você estiver analisando o tempo que os clientes passam em seu site de e-commerce antes de fazer uma compra, os dados seriam mais contínuos (por exemplo, minutos). Um gráfico de densidade poderia ser usado para mostrar a distribuição do tempo de navegação, ajudando a identificar se a maioria dos clientes compra rapidamente ou passa muito tempo navegando antes de decidir.

E aí, o que achou da resposta do ChatGPT?

O importante é perceber que ele nos deu uma boa explicação sobre a diferença e até apontou exemplos em que podemos escolher entre um tipo de histograma ou outro a depender da natureza de nossos dados.



## Distribuindo dados por característica

No vídeo anterior, iniciamos com mais um gráfico voltado à distribuição dos dados. Agora, vamos escalar um pouco mais essa parte da distribuição dos dados, inserindo uma nova variável aqui dentro.

E qual seria essa variável? Vamos adicionar algo categórico à nossa análise. Além disso, vamos ver também se o ChatGPT consegue responder bem a essa inserção de mais uma variável para construir uma distribuição.

Vamos, então, à leitura do nosso questionamento:

7º Questionamento: Como está distribuído o nosso público por idade e sexo biológico? Existe um grupo que sobressai?

Então, aqui ele quer saber a distribuição por duas variáveis. Uma variável numérica, que seria a idade, e uma variável categórica, que é sexo biológico, podendo ser masculino e feminino no exemplo da nossa base de dados.

Para isso, usaremos o seguinte prompt:


Temos um DataFrame chamado "df" e queremos construir um visual que apresente a distribuição do nosso público por idade e sexo biológico.

Em "df", temos as colunas:

- idade: idade da pessoa que realizou a compra
- sexo_biologico: sexo biológico da pessoa que realizou a compra, com dados categóricos

Desejamos criar uma visualização que apresente a distribuição desses dados.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Quando falo "com dados categóricos", é porque variáveis que são categóricas podem vir tanto no formato de texto,como está no nosso exemplo, quanto como números. Poderia ser zero e um, zero representando um sexo e 1 representando o outro sexo, por exemplo.

Então, vamos rodar esse nosso prompt aqui e trazer o código que vai gerar a visualização.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Supondo que você tenha o DataFrame "df" carregado

# Visualização em gráfico de barras empilhadas
plt.figure(figsize=(12, 8))
sns.histplot(data=df, x='idade', hue='sexo_biologico', multiple='stack', bins=20, palette='Set2')
plt.title('Distribuição do Público por Idade e Sexo Biológico')
plt.xlabel('Idade')
plt.ylabel('Contagem')
plt.legend(title='Sexo Biológico')

plt.show()
````

Então, esse é o nosso código para gerar o gráfico que pedimos. Ele traz a mesma importação das bibliotecas, com a diferença que ele traz a Seaborn. Então, vamos tirar essa parte do Matplotlib e o Pandas, e aquela parte também do comentário, do df.

Vamos manter o Seaborn, porque é a primeira vez que estamos trabalhando com ele aqui dentro do nosso projeto.

Ele vai criar uma visualização de barras empilhadas, ele já está dizendo aqui, em que ele vai ter um plt.figure, que é um fig size de 12 por 8. Vamos reduzir já para 10 por 6, para ficar num tamanho bom para visualizar.

````python
# Visualização em gráfico de barras empilhadas
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='idade', hue='sexo_biologico', multiple='stack', bins=20, palette='Set2')
plt.title('Distribuição do Público por Idade e Sexo Biológico')
plt.xlabel('Idade')
plt.ylabel('Contagem')
plt.legend(title='Sexo Biológico')

plt.show()
````

Ele vai fazer um sns.histplot, ou seja, ele vai criar um histograma de barras empilhadas. E, para isso, ele vai passar como dado o df, o eixo x, que é a nossa parte de baixo aqui, que são as coordenadas de idade, e o eixo y vai ser justamente a contagem, já que estamos trabalhando com histograma. E para as cores, que é o hue, esse significa tonalidade, ou seja, as cores vão ser dadas pelo valor de sexo biológico. Vai ter duas cores, uma para masculino e uma para feminino.

O multiple = 'stack' está dizendo que vamos colocar empilhados os nossos dados, então sempre que tivermos mais variáveis, ele vai empilhar. Se tivermos, por exemplo, 30 anos masculino e 30 anos feminino, ele vai empilhar cada um desses, para que ele mantenha os dois dados no mesmo intervalo.

O bins vai ser de 20, os intervalos.

No fim, estamos colocando uma paleta de cores pastel, de um tom pastel. Vamos passar o título "Distribuição do público por idade e sexo biológico", o x vai ser idade, o y vai ser contagem, e a legenda vai ter "sexo biológico".

Vamos rodar aqui esse gráfico para visualizar a nossa resposta.

Note que ele já trouxe, no eixo x, a nossa idade. E conseguimos notar que tem duas cores aqui, uma representando masculino e feminino. Mas a legenda não conseguiu trazer o que cada cor está indicando. E além disso, é pouco prático esse gráfico, não é mesmo? Porque por mais que tenhamos o valor de contagem, não conseguimos dizer se a barra de cima é menor do que a de baixo e vice-versa.

Então, por mais que consigamos ter a noção das contagens, não temos noção de quem tem mais e quem tem menos, dentro daquele intervalo. Além de trazer dois valores juntos, empilhados, o visual ficou um pouco poluído.

Então, vamos pedir para o ChatGPT mudar, trocar o tipo de visual, para que consigamos aplicá-lo aqui. Nesse exemplo, vai ser bem legal trazer um gráfico, por exemplo, de um BoxPlot, porque ele vai criar duas caixas separadas, que vão trazer os dados para masculino e os dados para feminino. Vamos fazer esse pedido para o ChatGPT.

Usaremos o seguinte prompt:

     Acho que podemos melhorar esse gráfico. Que tal mudar para um boxplot?

Simplesmente só isso. Não vamos passar mais nada, só vamos dizer que queremos trazer no formato BoxPlot. Vamos rodar esse prompt e trazer o novo código, fazendo essas operações.

Esse foi o código gerado:

````python
# Visualização em boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='sexo_biologico', y='idade', palette='pastel')
plt.title('Distribuição da Idade por Sexo Biológico')
plt.xlabel('Sexo Biológico')
plt.ylabel('Idade')

plt.show()
````

Vamos manter só a parte do Seaborn. Na verdade, vamos tirar porque já temos ele exibido lá em cima, então, vamos tirar essa parte. E aqui vamos ter uma geração do BoxPlot, ele já vai avisando. Vamos usar o tamanho de 10 por 6.

O BoxPlot vai receber o dado df. E o valor de X receberá o sexo biológico. Ou seja, no eixo X, vamos ter a separação masculino e feminino, e no eixo Y, vamos ter a idade.

Não vamos ter mais a contagem no lado esquerdo. Vamos ter agora o valor de idade. O gráfico vamos dizer que é a "Distribuição da idade do sexo biológico", o X label "sexo biológico", Y label "idade". E vamos fazer um plt.show para exibir esse nosso gráfico.

Os pontos que ficam acima, que seriam os outliers, seriam os pontos que passam de um limite, que tem um cálculo que é feito dentro do boxplot, que seriam valores que passam um pouco da média, do valor que seria ótimo aqui para os gráficos. Só que esses valores são representados aqui com as idades até 70 anos.

Ou seja, tem projetos que pode ser legal talvez tirar os outliers, mas o nosso não vai fazer sentido, porque vamos ter o nosso público variando de 18 anos até 70 anos. No feminino vai ter um pouquinho mais em relação ao masculino. Já esse traço no meio representa a mediana. O que seria a mediana? Se dividirmos os nossos dados, o meio vai ser esse valor, que provavelmente deve estar entre 38 a 37 anos. Vale observar para ver qual seria esse valor exato. E essa parte aqui mais acima da nossa caixa seria os 75 dados. 65% dos dados vão estar dentro desse ponto.

E claro, aqui, essa parte entre 25 e 75 é o que chamamos de intervalo interquartil, onde vão estar a maior parte dos nossos dados. Pense num gráfico de distribuição normal em que vamos ter aquela questão da sinusoidal, daquela onda em formato de sino. A maior parte dos dados ficam concentrados entre esse intervalo interquartil, entre 25% e 75% dos nossos dados.

# Faça como eu fiz: gráfico de árvore

Vamos construir um novo visual com os dados que estamos trabalhando no projeto do curso?

O visual que vamos trabalhar aqui é um gráfico de árvore, mais conhecido como Treemap. Ele é um tipo de visual muito utilizado para mostrar a composição e a hierarquia dos dados usando formas geométricas retangulares coloridas. Podemos observar na imagem abaixo esse visual:



image.png

Cada retângulo nesse visual tem o seu tamanho baseado no valor dos dados representados, ou seja, quanto maior o retângulo, mais vezes aquele valor aparece no conjunto de dados. Para o caso do faturamento total, quanto maior o faturamento, maior também será o tamanho dos retângulos.

As cores também são utilizadas para informar algo ao nosso público. Ela pode ser utilizada para representar uma variável categórica ou variável numérica, como regiões do país ou população das regiões, respectivamente.

## Este visual pode ser utilizado quando:

  - Queremos exibir grandes quantidades de dados hierárquicos.

  - Gráficos de setores ou barras não conseguem lidar com um grande volume de valores.

  - Desejamos mostrar as proporções entre cada parte e o todo.

  - Queremos detalhar os dados com o auxílio do tamanho e cores disponíveis.

Agora que aprendemos o que é e para que serve um Treemap, vamos construir o nosso para representar Faturamento Total por Região e Estado da Zoop.

## Construindo o visual

Vamos passar para o ChatGPT o seguinte prompt que além de enxugar os dados para criar uma tabela com apenas os valores que desejamos apresentar no visual, vai criar o visual de Treemap utilizando a biblioteca Plotly:

Temos um DataFrame chamado "df" e queremos construir um Treemap que apresente o faturamento total por região e estado.

Em "df", temos as colunas:

  - regiao: com as regiões dos(as) clientes. Com os valores de "Centro-Oeste", "Nordeste", "Norte", "Sudeste" e "Sul".

  - uf: com os estados dos(as) clientes.

  - faturamento: com os valores das vendas.

Crie um novo DataFrame chamado "vendas_2023" que filtre o df para manter apenas as colunas "regiao", "uf" e "faturamento" e crie um gráfico Treemap a partir desta base.

Para path passe apenas a coluna "uf", para values a coluna "faturamento" e para color a coluna "regiao". Para colorir cada região utilize as seguintes cores:

  - Centro-Oeste: "#0081a7"

  - Nordeste: "#00afb9"

  - Norte: "#fdfcdc"

  - Sudeste: "#fed9b7"

  - Sul: "#f07167"

Ajuste a largura (width =1400) e altura (height = 600) do Treemap e adicione os valores dentro dos retângulos. Utilize a biblioteca Plotly no Python para criar essa visualização.

O código gerado pelo prompt foi o seguinte:

````python
import pandas as pd
import plotly.express as px

# Supondo que você já tenha o DataFrame df
# Filtrar as colunas desejadas
vendas_2023 = df[['regiao', 'uf', 'faturamento']]

# Criar o gráfico Treemap
fig = px.treemap(vendas_2023, path=['regiao', 'uf'], values='faturamento', color='regiao',
                 color_discrete_map={'Centro-Oeste': '#0081a7',
                                     'Nordeste': '#00afb9',
                                     'Norte': '#fdfcdc',
                                     'Sudeste': '#fed9b7',
                                     'Sul': '#f07167'}
                 )

# Atualizando o layout para adicionar título
fig.update_layout(width=1400, height=600, title='Faturamento Total por Região e Estado', margin = dict(t=25, l=0, r=0, b=0))

# Adicionar os valores dentro dos retângulos
fig.update_traces(textinfo='label+value')

# Exibir o gráfico
fig.show()
````

image.png

Note que possuímos retângulos de diversos tamanhos apontando a representação de cada Estado no faturamento da Zoop. Este visual apresenta as proporções dos dados, identificando, pelas cores, quais Estados compõem a mesma região.

É importante enfatizar que, como essa visualização carrega diversas formas de destaque de dados (cores, áreas e sub-áreas), precisamos fazer uso dela com bastante cuidado, pois podemos adicionar um excesso de elementos que podem dificultar a análise dos dados de forma rápida e objetiva.

Se você seguiu o passo a passo, conseguirá visualizar o gráfico em seu notebook.

## Gráfico de colunas e barras horizontais

Um gráfico de colunas é ideal para representação de valores numéricos de uma ou mais categorias de dados, sejam elas discretas ou de intervalos definidos. Ele propicia uma comparação rápida e direta entre diferentes categorias, sendo um dos gráficos mais utilizados em relatórios e apresentações. O exemplo abaixo demonstra esse comportamento:



image.png

Já o gráfico de barras horizontais é ideal quando queremos apresentar as categorias escritas à esquerda do gráfico de forma horizontal, sendo a forma mais fluida de ler textos que possam ser extensos. O destaque deste tipo de gráfico está primeiro na categoria dos dados e, em sequência, nos valores numéricos.

image.png

## Gráfico de colunas agrupadas e colunas empilhadas

Um gráfico de colunas ou barras agrupadas é ideal quando desejamos comparar categorias diferentes em relação a uma variável, mostrando como os valores se comparam dentro de cada categoria (no exemplo, os trimestres) em colunas separadas (no exemplo, produto E, D e F) e entre os diferentes grupos, facilitando a visualização de padrões e diferenças entre grupos.

image.png

Um gráfico de colunas ou barras empilhadas é importante quando desejamos adicionar à comparação entre períodos ou categorias a comparação também entre os dados de cada período/categoria. Pense em uma lupa que “vasculha”, ou seja, que procura de modo minucioso cada coluna e observa a composição de cada uma delas de acordo com a variável desejada (no exemplo, as preferências de consumo de bebidas quentes: café e chá).

image.png

## Gráfico de linha

Os gráficos de linha são ideais para representar as tendências ao longo do tempo ou de acordo a uma sequência que possui uma ordenação. Quando possuímos uma grande quantidade de períodos para os nossos dados, vale a pena recorrer a gráficos de linha(s) para representar as séries temporais.

image.png

Os gráficos de linhas adicionam uma variável categórica às análises, permitindo comparação de séries temporais em curtos ou longos períodos de tempo. No exemplo abaixo, observamos o faturamento por região dividido pelos trimestres de um ano qualquer. É possível comparar os faturamentos em cada região em diferentes momentos utilizando apenas um visual.

image.png

## Gráfico de pizza e rosca

O gráfico de pizza é utilizado para representar as proporções de diferentes categorias ou partes de um todo. É ideal para composição estática dos dados seja por valores numéricos ou percentuais. É preciso bastante cuidado em sua utilização, evitando mais de 5 fatias (categorias) e também o uso de fatias muito próximas (por exemplo 51% e 49%) ou muito distantes entre si (por exemplo, 99% e 1%).

image.png

Um gráfico de setor alternativo ao gráfico de pizza é o gráfico de rosca, ele é adequado para quando queremos reduzir a área colorida para representar as fatias, passando com coroas circulares (aros). Este gráfico e o de pizza são ideais para poucas categorias de dados, se possível, dados dicotômicos (2 categorias, como no gráfico abaixo). Eles representam a proporção dos dados de maneira estática.

image.png

## Gráfico de Cascata

Gráficos de cascata são comumente utilizados para visualizar ajustes incrementais a um valor inicial, tornando-os ideais para visualizar informações e flutuações dos valores numéricos ao longo do tempo ou de uma sucessão de etapas. No exemplo abaixo, temos a representação da Receita Bruta e as deduções (Despesas) até a Receita Líquida da Serenatto Café & Bistrô.

image.png

## Histograma

Os histogramas são utilizados para observar a distribuição dos dados em um dado intervalo. Esses intervalos, que também são chamados de “bins”, apresentam quantas observações estão presentes naquela faixa de dados. Os histogramas também auxiliam na identificação de tendências centrais, variabilidade dos dados e possíveis outliers (valores atípicos).

image.png

## Gráfico de dispersão

Os gráficos de dispersão são utilizados para representar a relação ou distribuição entre duas variáveis numéricas. Cada ponto no gráfico representa a combinação entre as coordenadas das duas variáveis, que também chamamos de observação. Eles são utilizados largamente em ciência de dados, pois são visualizações que auxiliam na identificação de padrões, tendências e correlações entre as variáveis que são de extrema importância para o aprendizado de máquina (ML).

Podemos utilizar o seguinte prompt para a resolução do 1º desafio:

Temos um DataFrame chamado "df" e queremos construir um visual que apresente a proporção dos(as) usuários(as) que realizaram o pré-cadastro na Zoop Pay, nova solução da Zoop.

Em "df", temos a coluna "zoop_pay" que dispõe de tais dados. Construa um gráfico de rosca em que o valor "Sim" receba a cor #2a9d8f e "Não" a cor #edede9.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Utilizando a resposta do ChatGPT, obtemos o seguinte código:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Contabiliza a quantidade de usuários que realizaram o pré-cadastro na Zoop Pay
pre_cadastro_counts = df['zoop_pay'].value_counts()

# Define as cores para cada valor
colors = ['#2a9d8f' if label == 'Sim' else '#edede9' for label in pre_cadastro_counts.index]

# Cria o gráfico de rosca
plt.figure(figsize=(8, 6))
plt.pie(pre_cadastro_counts, labels=pre_cadastro_counts.index, autopct='%1.1f%%', startangle=140, colors=colors, wedgeprops=dict(width=0.3))

# Adiciona um título ao gráfico
plt.title('Proporção de Pré-cadastro na Zoop Pay')

# Exibe o gráfico
plt.show()
````

Temos um DataFrame chamado "df" e queremos construir um visual que apresente a distribuição das idades dos(as) clientes por região.

Desejamos criar uma visualização de boxplot que apresente a distribuição desses dados. Coloque as idades no eixo x e as regiões no eixo y

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Utilizando a resposta do ChatGPT, obtemos o seguinte código:

# Para saber mais: design de gráficos

Nesta aula, vamos conversar bastante sobre a criação dos visuais do curso e exercitar técnicas para trabalhar a estética dos gráficos. Dessa forma, nada melhor que conferir três propriedades fundamentais do design para garantir uma boa comunicação visual. Vamos lá?

## 1. Uso das cores

O uso correto das cores em projetos de natureza visual é importante para garantir legibilidade e uniformidade aos materiais e, para isso, dois parâmetros devem ser observados: contraste e consistência.

## Contraste

O contraste nada mais é do que a diferença entre duas coisas, num contexto comparativo. Então, na prática, avaliamos o contraste relacionando cores, já que ele não faz sentido numa cor sozinha. Contrastes mais fortes significam cores mais diferentes entre si, enquanto contrastes mais fracos denotam cores mais parecidas - aqui um ponto importante é ter cautela com essa aproximação porque cores similares dificultam o entendimento dos gráficos.

image.png

Compare os dois gráficos acima. Note como no da esquerda o contraste das cores é menor e, com isso, a separação entre as partes da pizza é menos visível. Usar uma única cor como guia não é problemático, mas se atente em trazer contraste suficiente para todos os tons escolhidos.

## Consistência

Em um projeto com muitos gráficos deve existir um padrão no uso das cores. Do contrário, os materiais não parecem fazer parte do mesmo conjunto e passam a sensação de amadorismo na sua construção, transmitindo uma ideia de que eles foram feitos sem seguir um padrão bem definido.

image.png

Note como os gráficos à direita, que seguem uma mesma orientação cromática, tem um acabamento mais profissional.

## 2. Tipografia

Os gráficos conferem um arranjo mais visual às informações textuais e trazem uma disposição mais clara para os dados expostos do que texto corrido. Ainda assim, os gráficos serão acompanhados de alguma parcela de texto, como títulos e legendas, e eles precisam ser legíveis. Para isso, dois atributos relacionados à tipografia devem ser considerados:

Tamanho: as palavras têm que ter um tamanho mínimo para que sejam lidas sem maiores dificuldades. Em trabalhos impressos, use pelo menos 10 pt e 20 pt para gráficos em apresentações.

Estilo: evite tipografias muito fantasiosas, que possuam um perfil cursivo ou com características excessivamente acentuadas, porque elas tendem a ter legibilidade pobre em dimensões menores e em extensões maiores de texto.

image.png

Na dúvida, dê prioridade às famílias tipográficas mais simples, como no gráfico à direita.

## 3. Hierarquia da informação

Já se viu numa conversa em que todo mundo fala ao mesmo tempo e ninguém se entende? O mesmo acontece na comunicação visual se tudo que está na tela ou no papel quer ser destaque ao mesmo tempo: é preciso definir apenas um ou dois elementos. Deste modo, cores e texto, especialmente, têm que ser manipulados de maneira a não saturar o leitor de informações

image.png

Evite selecionar muitas cores “fortes” e não exagere no tamanho e no peso do texto. Repare como o gráfico da esquerda é muito mais “poluído” visualmente, em que todos os elementos “falam” ao mesmo tempo. Procure balancear mais esses elementos visuais, como feito no gráfico à direita, onde cores e tipografia foram retrabalhados.

Agora chegou a hora de refinarmos as nossas visualizações. Vamos começar com o 1º gráfico que fizemos, aquele com os métodos de pagamento?

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA1, as colunas na cor VERMELHO1 e o tamanho da figura deve ser de (10, 6).

Código inicial gerado pelo ChatGPT

import pandas as pd
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Identidade visual Zoop
CINZA_1 = "#E6E6E6"
VERMELHO_1 = "#FF4D4F"
CINZA_5 = "#555555"
AZUL_1 = "#1890FF"

# Agrupar por método de pagamento e contar o total de ocorrências
metodos_de_pagamento = df['metodo_pagamento'].value_counts()

# Configurações do gráfico
plt.figure(figsize=(10, 6), facecolor=CINZA_1)
ax = plt.axes()
ax.set_facecolor(CINZA_1)
plt.bar(metodos_de_pagamento.index, metodos_de_pagamento, color=VERMELHO_1)

# Adicionar texto com o total acima de cada barra
for i, v in enumerate(metodos_de_pagamento):
    plt.text(i, v + 0.1, str(v), ha='center', va='bottom', color=AZUL_1, fontsize=12)

# Configurações adicionais
plt.title('Métodos de Pagamentos mais Utilizados em 2023', fontsize=18, color=CINZA_5)
plt.xticks(fontsize=12, color=AZUL_1)
plt.yticks([])
plt.xlabel('Método de Pagamento', fontsize=12, color=AZUL_1)
ax.spines[['top', 'right', 'left']].set_visible(False)

plt.show()

Não precisamos fazer a importação das bibliotecas Pandas e Matplotlib, porque já fizemos anteriormente no nosso projeto, então podemos remover essas importações do código.

Também não precisamos do trecho que define as cores para as variáveis passadas, porque já as definimos na paleta, então vamos deletá-lo do código para que as variáveis não sejam sobrescritas.

image.png

## Entendendo o código gerado pelo GPT

Nesse código, colocamos um facecolor cinza dentro do plt.figure ( ), o que que colocou a cor cinza no fundo da figura. Também colocamos a cor cinza no fundo do plot em si usando o ax.set_facecolor( ). Mudamos a cor do plt.bar também, que é o gráfico em si, para vermelho, para representar as colunas.

Também fizemos adição de texto acima das colunas com o total de cada método de pagamento, na posição i, v. O i é o índice da barra (como barra 0, barra 1, barra 2 e barra 3) retornado pelo enumerate, no eixo X, e v é o valor de Y mesmo associado a cada barra, que aparece acima delas.

Esses valores vão ser representados com um offset de 0.1, ou seja, estamos colocando um pouco acima da barra para não ficar escrito exatamente em cima dela. E com o str(v), estamos colocando esse valor no formato de string, ou seja, no formato de texto. Afinal, o valor é um inteiro, então precisamos transformar em texto para representá-lo.

Com ha='center' alinhamos o texto ao centro da barra, com o va='bottom' alinhamos o texto verticalmente à base (parte inferior) do texto, depois definimios a cor do texto como AZUL_1 e tamanho 12.

Nos ajustes adicionais, configuramos o título em tamanho 18 e cor CINZA_5. Passamos os xticks, que são os rótulos de categoria de pagamento, com fonte 12 e a cor azul; no yticks passamos uma lista vazia para retirar o ticks, ou seja, não exibir a escala de números do eixo Y. No xlabel, que é o rótulo do eixo X em si, também passamos fonte 12 e cor AZUL_1.

Por fim, aplicamos o ax.spines[['top', 'right', 'left']].set_visible(False), que retira as bordas do topo, da esquerda e da direita. Ou seja, não temos mais aqueles traços que passam em volta do nosso plot, mantendo apenas o que fica na parte inferior, no eixo X.

# Para saber mais: elementos de um gráfico no Matplotlib

No vídeo anterior, criamos um prompt que alterou uma série de elementos dentro do nosso visual, dando uma nova aparência a ele, seguindo as orientações na criação de um visual da Zoop em termos de cores, rótulos, títulos, dados entre outros. Vamos utilizar o próprio ChatGPT para responder estas dúvidas.

## Prompt:

Crie um texto que fale brevemente sobre os elementos de um gráfico na biblioteca Matplotlib como, por exemplo, xticks( ), yticks( ), xlabel( ), ylabel( ), text( ), annotate, entre outros. Dê exemplos para cada boa prática voltados a um projeto simples de Data Science com Python.

A resposta dada pelo ChatGPT é a seguinte:

A biblioteca Matplotlib é uma das ferramentas mais populares e poderosas para criação de gráficos e visualizações de dados em Python. Ela oferece uma vasta gama de funcionalidades e elementos que podem ser personalizados para melhorar a legibilidade e a apresentação das visualizações de dados. Aqui estão alguns dos elementos-chave que você pode utilizar ao criar gráficos com Matplotlib, acompanhados de boas práticas e exemplos voltados a um projeto simples de Data Science.

## 1. xticks e yticks
Essas funções são usadas para personalizar os ticks dos eixos X e Y. Ajustar esses elementos pode melhorar a legibilidade dos gráficos, especialmente quando se trabalha com dados que têm escalas grandes ou pequenas, ou quando os valores são densamente distribuídos. Exemplo: plt.xticks(range(0, 101, 10)) define os ticks do eixo X para valores de 0 a 100, com intervalos de 10.

## 2. xlabel e ylabel
São rótulos para os eixos X e Y, respectivamente. Essenciais para indicar o que cada eixo representa. Nomeie seus eixos com rótulos claros e informativos. Isso torna o gráfico compreensível sem necessidade de consultar o contexto externo. Exemplo: plt.xlabel('Tempo (anos)') e plt.ylabel('População (milhões').

## 3. Título (title) e subtítulos
Adiciona um título ao gráfico, fornecendo uma visão geral imediata do que o gráfico está apresentando. Subtítulos podem ser usados para fornecer informações adicionais ou contextuais. Use um título descritivo que possa diminuir a necessidade dos labels e considere subtítulos para detalhes adicionais, como a fonte dos dados ou observações específicas. Exemplo: plt.title('Crescimento Populacional ao Longo do Tempo').

## 4. Legenda (legend)
A legenda ajuda a identificar diferentes séries ou categorias dentro de um gráfico. É crucial quando um gráfico contém múltiplas linhas, barras, ou outros elementos visuais. Use legendas claras e posicione a legenda de forma a não obstruir os dados. plt.legend(loc='best') automaticamente escolhe a melhor localização para a legenda. Exemplo: plt.legend(['Produto A', 'Produto B']).

## 5. Anotação (text e annotate)
A anotação permite adicionar textos explicativos em pontos específicos do gráfico, o que pode ser útil para destacar características ou pontos de dados importantes. text permite adicionar texto em qualquer parte do gráfico, enquanto annotate é usado para adicionar anotações com a opção de incluir uma seta apontando para uma parte específica do gráfico. Use plt.annotate( ) para adicionar anotações que esclareçam pontos de dados incomuns ou tendências notáveis. Especifique a localização da anotação e o ponto de dados a ser anotado. Exemplo: plt.annotate('Pico Máximo', xy=(2020, 500), xytext=(2021, 600), arrowprops=dict(facecolor='black')) destaca um pico no gráfico.

## 6. Grade (grid)

Adiciona uma grade ao gráfico, o que pode melhorar significativamente a legibilidade ao estimar valores. Linhas de grade ajudam a ler os gráficos ao fornecer linhas de referência que facilitam a correlação dos valores dos dados com os eixos. Ative a grade quando tiver muitos pontos de dados ou quando precisar facilitar a comparação de valores. Exemplo: plt.grid(True) habilita a exibição da grade no gráfico.

Utilizar esses elementos adequadamente pode significativamente melhorar a qualidade das suas visualizações de dados, tornando-as não apenas mais atraentes visualmente, mas também mais informativas e fáceis de entender. Em projetos de Data Science, onde a comunicação eficaz dos resultados é essencial, dominar esses elementos do Matplotlib é uma habilidade valiosa.

## Vamos realizar essas práticas neste vídeo.

Prompt para adicionar um texto informativo
Já temos salvo no histórico do ChatGPT o último gráfico gerado. Vamos deixar claro para ele que queremos utilizar esse gráfico e fazer alterações nele. Para isso, vamos enviar o seguinte prompt para gerar o próximo plot:

O gráfico ficou muito bom! Agora quero adicionar um texto dentro da área do plot, vamos lá?

Para isso, some os valores das duas primeiras colunas, divida pela soma de totas as colunas e passe para o formato de porcentagem armazenando na variável "percentual"

Agora, na posição x = 2 e y = 2500, adicione o seguinte texto com tamanho 10 e cor AZUL_3: f"$\\bf{percentual:.2f}$% dos clientes utilizam Cartão de Crédito ou PIX \npara pagamento. Seria interessante recorrer a parcerias\ncom bancos para a construção de um Zoop Pay a fim\nde fidelizar a nossa clientela.".

Com esse prompt, estamos primeiramente contextualizando o gráfico e chamando o ChatGPT para modificá-lo.

Depois, estamos fazendo um cálculo. Temos diferentes valores para as quantidades de pagamentos realizados com cartão de crédito, Pix, cartão de débito e boleto. Queremos somar os valores de cartão de crédito e Pix, que são os dois maiores, dividir esse resultado pela soma de todos os pagamentos e multiplicar por cem.

Com isso, vamos descobrir o percentual de pagamentos via cartão de crédito e Pix em relação ao total de pagamentos. Por fim, armazenamos o resultado numa variável que chamamos de percentual.

No último parágrafo, indicamos uma posição e estilo para o texto e usamos um artifício para adicionar uma variável nesse texto e deixá-la em negrito: $\\bf{ }$.

Quando colocamos um texto no Python, só conseguimos deixá-lo normal ou em negrito, passando essa informação via parâmetros do texto. Só que não conseguimos deixar uma única variável em negrito se tivermos um texto normal ao lado dele — ou o texto é todo em negrito ou o texto é todo normal.

Para isso, utilizamos essa expressão chamada de MathTeX, um texto utilizado para fórmulas matemáticas no Python. Com esse artifício, estamos transformando o valor armazenado em percentual para negrito, formatando-o com duas casas decimais com 2f e, no final, colocamos o sinal de porcentagem: $\\bf{percentual:.2f}$%. O sinal de porcentagem não vai estar em negrito, pois está fora da variável.

No texto, falamos que esse percentual de clientes (armazenado na variável percentual) utiliza Cartão de crédito ou PIX para pagamento, e que seria interessante recorrer a parcerias com bancos para a construção de um "Zoop Pay", a fim de fidelizar a clientela da ZOOP. Esse é o texto que queremos exibir dentro da área útil do nosso gráfico.

## Códigos de cálculo de percentual e inserção de texto informativo

Vamos rodar o prompt no ChatGPT para gerar o código que gera esse visual. Ele nos devolve:

````python
# Agrupar por método de pagamento e contar o total de ocorrências
metodos_de_pagamento = df['metodo_pagamento'].value_counts()

# Calcular percentual
total_clientes = metodos_de_pagamento.sum()
percentual = (metodos_de_pagamento[0] + metodos_de_pagamento[1]) / total_clientes * 100

# Configurações do gráfico
plt.figure(figsize=(10, 6), facecolor=CINZA_1)
ax = plt.axes()
ax.set_facecolor(CINZA_1)
plt.bar(metodos_de_pagamento.index, metodos_de_pagamento, color=VERMELHO_1)

# Adicionar texto com o total acima de cada barra
for i, v in enumerate(metodos_de_pagamento):
    plt.text(i, v + 0.1, str(v), ha='center', va='bottom', color=AZUL_1, fontsize=12)

# Adicionar texto dentro da área do plot
plt.text(2, 2500, f"$\\bf{percentual:.2f}$% dos clientes utilizam Cartão de Crédito ou PIX \npara pagamento. Seria interessante recorrer a parcerias\ncom bancos para a construção de um Zoop Pay a fim\nde fidelizar a nossa clientela.",
            ha='left', va='center', color=AZUL_3, fontsize=10)

# Configurações adicionais
plt.title('Métodos de Pagamentos mais Utilizados em 2023', fontsize=18, color=CINZA_5)
plt.xticks(fontsize=12, color=AZUL_1)
plt.yticks([])
plt.xlabel('Método de Pagamento', fontsize=12, color=AZUL_1)
ax.spines[['top', 'right', 'left']].set_visible(False)

plt.show()
````

image.png

O texto contém a porcentagem calculada com a fórmula de percentual, resultando em 82,59% de clientes que usam cartão de crédito e PIX para realizar pagamentos na ZOOP.

O que trouxe esse valor para nós? Vamos recapitular.

No trecho de cálculo, criamos uma variável total_clientes que armazena o somatório de todos os pagamentos, considerando todos os métodos de pagamento. Depois criamos a variável percentual que armazena a soma dos dois maiores valores, ou seja, dos métodos de pagamento 0 e 1 — Cartão de crédito e PIX, respectivamente — dividida pelo total dos métodos de pagamento e multiplicado por 100. Isso traz o dado já em percentual mesmo para nós.

O trecho que adiciona um texto na área do plot usa o plt.text( ), que recebe as posições X e Y do texto, o texto formatado, a sua centralização na caixa de texto (invisível para nós) e o estilo.

Ou seja: fizemos apenas duas pequenas alterações no código do nosso gráfico para adicionar esse texto informativo.

Utilizamos uma das áreas úteis do nosso gráfico para trazer uma informação relevante, um insight a partir da leitura dos dados apresentados nesse gráfico.

Esse tipo de adição é interessante para trazer destaques ou informações sobre determinados dados, dentro do mesmo visual!

## Gráfico inicial

Para esse processo, vamos escolher um novo gráfico, que seria o gráfico de linha. Esse foi o terceiro gráfico que criamos, acerca das Vendas totais mensais da ZOOP. Vamos aplicar algumas técnicas nele, utilizando a IA do ChatGPT.

Inicialmente, vamos rodar o código que já tínhamos criado para a visualização original do nosso gráfico de linha:

````python
import pandas as pd
import matplotlib.pyplot as plt

# Dicionário de tradução de meses
meses = {
    'January': 'Jan',
    'February': 'Fev',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov',
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] = df['data'].dt.strftime('%B')
vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6))
plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color='#1890FF', linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais da Zoop', fontsize=18, color='#555555')
plt.xlabel('Mês', fontsize=12, color='#1890FF')
plt.ylabel('Vendas', fontsize=12, color='#1890FF')

# Alterar rótulos do eixo x utilizando o dicionário de tradução
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=10, color='#1890FF')

plt.show()
````

Se temos meses nos rótulos eixo X, sabemos que trata-se de um gráfico de tempo, e os valores do eixo Y estão expressos em reais, então sabemos que trata-se de faturamento de dinheiro.

````python
# Dicionário de tradução de meses
meses (
    'January': 'Jan',
    'February': 'Fev",
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    "June': 'Jun',
    'July': 'Jul,
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov,
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] df['data'].dt.strftime('%8')
vendas_mensais df.groupby('mes') ['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6), facecolor-CINZA_1)
ax plt.axes()
ax.set_facecolor(CINZA_1)
plt.plot(vendas_mensais.index, vendas_mensais, marker'o', color-AZUL_1, linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais em 2023', fontsize=18, color=CINZA_5)
plt.xlabel('')
plt.ylabel('')

# Formatando rótulos do eixo y em "R$ valor M"
formatted_yticks = ['R$ (:.2f} M'.format(val / 1e6) for val in ax.get_yticks()]
ax.set_yticks(ax.get_yticks()) # para evitar o Warning
ax.set_yticklabels (formatted_yticks, fontsize=12, color=AZUL_1)

# Configurações adicionais
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=12, color=AZUL_1)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Exibir o gráfico
plt.show()
````

Essa variável vai dividir esse valor por um milhão. Ou seja, se o dado é dois milhões, ela vai dividir por um milhão e vai dar dois. E esse dois vai ser representado com .2f, duas casas decimais, no formato R$ (em reais).

Se levantar algum warning para você quando estiver executando o código, é só inserir o trecho ax.set_yticks(ax.get_yticks()) para resolver. Isso acontece porque quando queremos formatar os nossos ticks, ele reconhece que a quantidade dos ticks pode ter mudado, porque você está fazendo uma alteração manual nos valores.

Então você tem que garantir que a quantidade de variáveis que você está mudando dentro dos ticks é a mesma que você tem, porque senão vai ter uma confusão dentro do código e isso pode resultar que os dados não fiquem posicionados de maneira correta. É um ajuste para garantir a funcionalidade correta na criação do seu gráfico.

Por fim, ele faz o set_yticklabels, que é inserir esses rótulos de Y no nosso gráfico, utilizando o formatted_yticks, no tamanho 12 e cor azul.

O plt_xticks é a mesma operação, e está inserindo apenas um tamanho 12 e uma cor azul para os rótulos do eixo X.

## Destaques

Nossa segunda operação será a inserção de alguns elementos visuais, usando o seguinte prompt:

Bacana! Agora que o nosso gráfico já está com a identidade visual, vamos adicionar alguns elementos visuais.

Desenhe 4 círculos de raio 200, transparentes com a borda na cor VERMELHO_1 e com o centro nas coordenadas do 5º, 8º, 11º e 12º pontos, respectivamente. No lado esquerdo de cada círculo enumere de 1 a 4.

Ou seja, os elementos visuais que queremos inserir são 4 círculos destacando os meses de maio, agosto, novembro e dezembro, enumerando esses quatro pontos.

Com isso, daremos foco para determinados dados não só circulando os pontos, mas também colocando um número neles para poder adicionar uma informação ou observação em legenda.

Vamos enviar esse prompt para o ChatGPT trazer o código gerado a partir dele para o nosso projeto:

````python
# Dicionário de tradução de meses
meses (
    'January': 'Jan',
    'February': 'Fev",
    'March': 'Mar',
    'April': 'Abr',
    'May': 'Mai',
    "June': 'Jun',
    'July': 'Jul,
    'August': 'Ago',
    'September': 'Set',
    'October': 'Out',
    'November': 'Nov,
    'December': 'Dez'
}

# Agrupar por mês e calcular as vendas totais
df['mes'] df['data'].dt.strftime('%8')
vendas_mensais df.groupby('mes') ['faturamento'].sum().reindex(meses.keys())

# Configurações do gráfico
plt.figure(figsize=(12, 6), facecolor-CINZA_1)
ax plt.axes()
ax.set_facecolor(CINZA_1)
plt.plot(vendas_mensais.index, vendas_mensais, marker'o', color-AZUL_1, linewidth=2)

# Adicionar rótulos e título
plt.title('Vendas Totais Mensais em 2023', fontsize=18, color=CINZA_5)
plt.xlabel('')
plt.ylabel('')

# Formatando rótulos do eixo y em "R$ valor M"
formatted_yticks = ['R$ (:.2f} M'.format(val / 1e6) for val in ax.get_yticks()]
ax.set_yticks(ax.get_yticks()) # para evitar o Warning
ax.set_yticklabels (formatted_yticks, fontsize=12, color=AZUL_1)

# Configurações adicionais
plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=12, color=AZUL_1)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Adicionar círculos numerados
for i, point in enumerate([(4, vendas_mensais.iloc[4]), (7, vendas_mensais.iloc[7]), (10, vendas_mensais.iloc[10]), (11, vendas_mensais.iloc[11])], start=1):
    plt.scatter(point[0], point[1], s=200, color'none', edgecolors=VERMELHO_1, alpha=0.7)
    plt.text(point[0] - 0.3, point[1] + 0.1, str(i), fontsize=12, color=VERMELHO_1)

# Exibir o gráfico
plt.show()
````

image.png

Colocamos um círculo seguido dos números nos pontos desejados. O trecho do código que faz essa modificação do código é o # Adição de círculos numerados.

Esse código está gerando um laço for que vai receber um valor i e um valor point. O point é o ponto, as coordenadas desses círculos, e o i vai ser o índice desses círculos (de 1 a 4).

O point vem de uma lista de tuplas, e cada tupla vem com um par ordenado. O X é a posição em X no gráfico: 4, 7, 10 e 11, que seriam justamente os meses de maio, agosto, novembro e dezembro. Por que iniciamos em 4 e não 5? Porque no Python a contagem começa do zero, então estamos sempre um mês "atrasados" na representação da posição em X.

O Y vai ler justamente os valores que estão naquele eixo. No caso, os valores de cada mês vão ser lidos de vendas_mensais a partir do iloc, porque ele faz a localização pelo índice. Então, passamos essas coordenadas X e Y.

Para isso ele vai criar dois desenhos: um scatter e um text. O scatter é um gráfico de pontos, que vai passar por o ponto zero, que seria a coordenada X do gráfico, e o ponto um, que é a coordenada Y. Depois, ele vai criar um ponto de tamanho de 200, com nenhuma cor (transparente) e cor de borda vermelha. O alpha com valor 0.7 garante certa transparência nesse círculo.

Por fim, vamos ter um plt.text qie vai pegar essas coordenadas X = 0 e Y = 1 de 0, e adicionar um offset de - 0.3 em X para puxar o texto um pouco para o lado e + 0.1 em Y para puxar para cima, para ele não ficar em cima do ponto do gráfico. Além disso, passamos esse valor em formato string, tamanho 12 e cor vermelha.

Se espelharmos esse gráfico e compararmos com o inicial, notaremos a diferença. Removemos muito elementos sem sentido, como a representação dos valores no eixo Y. Sabíamos que eles estavam numa escala de milhões devido ao indicador "1e6", mas isso não é nada intuitivo, e também não deixamos nenhum indicativo de que esses valores são em reais. No novo gráfico, deixamos isso mais claro.

Retiramos também as bordas que limitavam o espaço livre do gráfico, e deixamos todas as cores de rótulos azuis, porque elas estavam diferentes no primeiro gráfico.

E, por fim, a alteração mais relevante foi o destaque dos dados em pontos mais altos no gráfico. O gráfico inicial mostra a evolução das vendas conforme os meses, mas no segundo gráfico chamamos atenção para quatro pontos específicos sobre os quais queremos adicionar uma informação.

Essas informações podem ser passadas via texto, abaixo do gráfico. Por exemplo, podemos tentar explicar a alta de vendas desses meses:

A propriedade da largura da rosca (wedgeprops) precisa ser de 0.4, a posição das porcentagens dentro da fatia (pctdistance) precisa ser de 0.8, o ângulo de início de 180 (startangle) e sem explosão de fatias.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Utilizando a resposta do ChatGPT, obtemos o seguinte código:

````python
import matplotlib.pyplot as plt

# Contabiliza a quantidade de usuários que realizaram o pré-cadastro na Zoop Pay
# pre_cadastro_counts = df['zoop_pay'].value_counts()

# Define as cores para cada valor
colors = [AZUL_3 if label == 'Sim' else CINZA_3 for label in pre_cadastro_counts.index]

# Cria o gráfico de rosca
plt.figure(figsize=(6, 6), facecolor=CINZA_1)
plt.gca().set_facecolor(CINZA_1)    # Altera a cor do fundo do plot
patches, texts, autotexts = plt.pie(pre_cadastro_counts, labels=pre_cadastro_counts.index, autopct='%1.1f%%', startangle=180,
                                    colors=colors, wedgeprops=dict(width=0.4), pctdistance=0.8)

# Adiciona o título com a cor e tamanho especificados
plt.title('Pré-cadastrado(a) na Zoop Pay?', color=CINZA_5, fontsize=18)

# Configura as cores e tamanhos dos rótulos das fatias e porcentagens
for text, autotext, color in zip(texts, autotexts, [CINZA_3, AZUL_3]):
    text.set_color(color) # Define a cor do rótulo da fatia igual à cor da fatia
    text.set_fontsize(12)
    autotext.set_color(BRANCO)  # Define a cor das porcentagens como branco
    autotext.set_weight('bold')  # Define o texto das porcentagens em negrito
    autotext.set_fontsize(12)

# Exibe o gráfico
plt.show()
````

## Trazendo alterações para dentro da função

Antes de seguirmos para o refinamento dos gráficos, traremos de volta os dois gráficos, passando-os para a função. Faremos isso, pois precisamos fazer com que o bloco seja reprodutível e ainda consigamos resgatá-lo em qualquer momento que desejarmos dentro do nosso projeto.

Então, no Colab, temos o código referente ao gráfico 1, que representa os métodos de pagamento mais utilizados de 2023. Nele, colocávamos um texto os valores nas barras. A única diferença é que agora colocamos como uma função, def métodos_pag( ) indentamos o código dentro dessa função.

No fim do código, ao invés de passar o plt.show, transformamos em um return plt, ou seja, estamos salvando a figura utilizando essa função. Quando a função for chamada, executará essa figura.

````python
def metodos_pag():
  # Agrupar por método de pagamento e contar o total de ocorrências
  metodos_de_pagamento = df['metodo_pagamento'].value_counts()

  # Calcular percentual
  total_clientes = metodos_de_pagamento.sum()
  percentual = (metodos_de_pagamento[0] + metodos_de_pagamento[1]) / total_clientes * 100

  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)
  plt.bar(metodos_de_pagamento.index, metodos_de_pagamento, color=VERMELHO_1)

  # Adicionar texto com o total acima de cada barra
  for i, v in enumerate(metodos_de_pagamento):
      plt.text(i, v + 0.1, str(v), ha='center', va='bottom', color=AZUL_1, fontsize=12)

  # Adicionar texto dentro da área do plot
  plt.text(2, 2500, f"$\\bf{percentual:.2f}$% dos clientes utilizam Cartão de Crédito ou PIX \npara pagamento. Seria interessante recorrer a parcerias\ncom bancos para a construção de um Zoop Pay a fim\nde fidelizar a nossa clientela.",
          ha='left', va='center', color=AZUL_3, fontsize=10)

  # Configurações adicionais
  plt.title('Métodos de Pagamentos mais Utilizados em 2023', fontsize=18, color=CINZA_5)
  plt.xticks(fontsize=12, color=AZUL_1)
  plt.yticks([])
  plt.xlabel('Método de Pagamento', fontsize=12, color=AZUL_1)
  ax.spines[['top', 'right', 'left']].set_visible(False)
  plt.grid(False)

  return plt
````

Executamos os dois códigos. Enquanto isso, no bloco abaixo temos o segundo gráfico de linhas referente as vendas totais mensais em 2023. As alterações traremos também para dentro de uma função, chamamos de def vendas_mensais( ) e o return plt para trazer o visual.

````python
def vendas_mensais():
  # Dicionário de tradução de meses
  meses = {'January': 'Jan', 'February': 'Fev', 'March': 'Mar', 'April': 'Abr',
           'May': 'Mai', 'June': 'Jun', 'July': 'Jul', 'August': 'Ago',
           'September': 'Set', 'October': 'Out', 'November': 'Nov', 'December': 'Dez'}

  # Agrupar por mês e calcular as vendas totais
  df['mes'] = df['data'].dt.strftime('%B')
  vendas_mensais = df.groupby('mes')['faturamento'].sum().reindex(meses.keys())

  # Configurações do gráfico
  plt.figure(figsize=(12, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)
  plt.plot(vendas_mensais.index, vendas_mensais, marker='o', color=AZUL_1, linewidth=2)

  # Adicionar rótulos e título
  plt.title('Vendas Totais Mensais em 2023', fontsize=18, color=CINZA_5)
  plt.xlabel('')
  plt.ylabel('')

  # Formatando rótulos do eixo y em "R$ valor M"
  formatted_yticks = ['R$ {:.2f} M'.format(val / 1e6) for val in ax.get_yticks()]
  ax.set_yticks(ax.get_yticks()) # para evitar o Warning
  ax.set_yticklabels(formatted_yticks, fontsize=12, color=AZUL_1)

  # Adicionar círculos numerados
  for i, point in enumerate([(4, vendas_mensais.iloc[4]), (7, vendas_mensais.iloc[7]), (10, vendas_mensais.iloc[10]), (11, vendas_mensais.iloc[11])], start=1):
      plt.scatter(point[0], point[1], s=200, color='none', edgecolors=VERMELHO_1, alpha=0.7)
      plt.text(point[0] - 0.3, point[1] + 0.1, str(i), fontsize=12, color=VERMELHO_1)

  # Configurações adicionais
  plt.xticks(vendas_mensais.index, [meses[mes] for mes in vendas_mensais.index], fontsize=12, color=AZUL_1)
  ax.spines['top'].set_visible(False)
  ax.spines['right'].set_visible(False)

  return plt
````

Isso acontece, pois estão em função. Fizemos apenas a definição das duas funções. Quando fizermos a chamada da função, eles vão aparecer na tela. Faremos esse exemplo no terceiro gráfico que alteraremos dentro do projeto.

Agora, vamos partir para a modificação do terceiro gráfico, que é o faturamento por categoria de produto em 2023. Para relembrar, trazemos o código que executamos para a criação do gráfico na primeira versão e vamos rodá-lo.

````python
import matplotlib.pyplot as plt

# Calcular o faturamento para cada categoria
df['faturamento'] = (df['preco_unitario'] * df['quantidade']) + df['frete']

# Agrupar por categoria e calcular o faturamento total
faturamento_por_categoria = df.groupby('categoria')['faturamento'].sum().reset_index()

# Ordenar por faturamento
faturamento_por_categoria = faturamento_por_categoria.sort_values(by='faturamento', ascending=False)

# Visualização
plt.figure(figsize=(10, 8))
plt.barh(faturamento_por_categoria['categoria'], faturamento_por_categoria['faturamento'], color='skyblue')
plt.title('Faturamento por Categoria')
plt.xlabel('Faturamento')
plt.ylabel('Categoria')

plt.show()
````

Vamos partir para mais um gráfico? Agora, vamos refinar o gráfico de faturamento por categoria. Lembre que esse gráfico foi de barras horizontais.

Copiamos o prompt, colamos no ChatGPT e assim temos como retorno o seguinte código:

````python
def vendas_tri_metodo():
  # Agrupar por trimestre e método de pagamento e calcular o faturamento total
  vendas_por_trimestre_metodo = df.groupby(['trimestre', 'metodo_pagamento'])['faturamento'].sum().unstack()

  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)

  # Plotar o gráfico de barras empilhadas
  vendas_por_trimestre_metodo.plot(kind='bar', stacked=True, color=[BRANCO, CINZA_3, AQUA_1, AQUA_3], ax=ax, edgecolor=CINZA_1)

  # Adicionar rótulos e título
  plt.title('Vendas por Trimestre e Método de Pagamento em 2023', fontsize=18, color=CINZA_5)
  plt.xlabel('')
  plt.ylabel('')
  plt.xticks(rotation=0, fontsize=12, color=AZUL_1)
  plt.yticks(ax.get_yticks(), ['R$ {:.0f} M'.format(val/1e6) for val in ax.get_yticks()], fontsize=12, color=AZUL_1)
  ax.spines['top'].set_visible(False)
  ax.spines['right'].set_visible(False)
  ax.spines['left'].set_visible(False)
  ax.spines['bottom'].set_visible(False)
  ax.yaxis.grid(color='gray', linestyle='--', linewidth=0.5)
  ax.set_axisbelow(True) # adicionar grid atrás

  return plt
````

Agora, vamos refinar o gráfico de rosca que produzimos para a proporção de usuários aderindo ao programa de cashback.

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA_1, a cor da fatia de valor "Não" igual a CINZA_3, a cor da fatia de valor "Sim" igual a AZUL_1 e o tamanho da figura deve ser de (6, 6).

O título da figura é "Participam do programa de cashback?" com tamanho 18 e sua cor CINZA_5.

Os rótulos das fatias precisam ter a mesma cor das fatias e tamanho 12 e as porcentagens precisam estar na cor BRANCO, também com tamanho 12.

A propriedade da largura da rosca (wedgeprops) precisa ser de 0.4, a posição das porcentagens dentro da fatia (pctdistance) precisa ser de 0.8, o ângulo de início de 180 (startangle) e sem explosão de fatias.

Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

Pedimos o ajuste da cor de fundo para cinza, para que a fatia "Não" renha a cor cinza 3 e a se "Sim" o azul 1 e que o tamanho da figura seja de 6x6. Definimos o título e informamos que as fatias e rótulos precisam ter as mesmas cores.

Solicitamos as porcentagens em uma cor branca dentro do visual das fatias, no tamanho 12. Além disso, ajustaremos as propriedades da rosca. Ao invés de 0.6, definimos como 0.4 para ficar mais fina. Definimos o ângulo que queremos iniciar a angulação em 180. Para isso, usaremos a biblioteca Python.

Ao rodar esse prompt no ChatGPT, temos a seguinte sugestão de código:

````python
def cashback():
  # Calcular a proporção de usuários aderindo ao programa de cashback
  proporcao_cashback = df['cashback'].value_counts(normalize=True)

  # Configurações do gráfico de rosca
  fig, ax = plt.subplots(figsize=(6, 6), facecolor=CINZA_1)
  ax.set_facecolor(CINZA_1)

  # Plotar o gráfico de rosca
  wedges, texts, autotexts = ax.pie(proporcao_cashback, labels=proporcao_cashback.index, autopct='%1.1f%%', startangle=180,
                                    wedgeprops=dict(width=0.4, edgecolor=CINZA_1), pctdistance=0.8, colors=[CINZA_3, AZUL_1])

  # Adicionar rótulos e título
  plt.title('Participam do Programa de Cashback?', fontsize=18, color=CINZA_5)

  # Configurar rótulos e porcentagens com as cores especificadas
  for text, autotext, color in zip(texts, autotexts, [CINZA_3, AZUL_1]):
      text.set_color(color)
      text.set_fontsize(12)
      autotext.set_color(BRANCO)
      autotext.set_fontsize(12)

  return plt
````

Isso é algo que precisa ter muita atenção, pois nem sempre o ChatGPT irá entender algumas interpretações e nesse caso precisamos deixar o mais claro e objetivo possível.

Agora, vamos comparar as duas versões desse gráfico. Quando analisamos o novo gráfico, percebemos primeiro a fatia do "Sim", pois a cor azul destaca muito mais que a cinza. Essa é uma das técnicas mais utilizadas no Storytelling com dados. Por meio das cores podemos estabelecer o foco que queremos que as pessoas tenham.

## Refinando o gráfico de avaliação dos produtos

Então, iniciaremos o processo para refinar os gráficos de distribuição, começando pelo gráfico de avaliação dos produtos. Rodamos o código do gráfico que criamos anteriormente.

````python
import matplotlib.pyplot as plt

# Supondo que você tenha o DataFrame "df" carregado

# Visualização em histograma
plt.figure(figsize=(10, 6))
plt.hist(df['avaliacao_compra'], bins=11, edgecolor='black', color='skyblue')
plt.title('Distribuição das Avaliações de Compra')
plt.xlabel('Avaliação')
plt.ylabel('Frequência')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
````

image.png

Nesse visual, temos um histograma de colunas que tem a distribuição, a frequência das avaliações dos clientes, de acordo com as compras realizadas.

Nele, aplicaremos um novo recurso que foca no texto e no visual. Você pode estar se perguntando, mas já não fizemos isso no gráfico de faturamento método de pagamento? Sim, fizemos algo similar inserindo só o texto. Porém, nesse caso, criaremos uma caixa de texto que irá focar em um ponto interessante no gráfico.

Para isso, usaremos o seguinte prompt:

Vamos agora para o gráfico da Distribuição das Avaliações de Compra que produzimos anteriormente. Vamos refiná-lo usando as seguintes orientações.

O nosso gráfico precisa seguir a identidade visual da Zoop, com o fundo da figura e do plot na cor CINZA_1, a cor das colunas VERMELHO_1 e o tamanho da figura deve ser de (10, 6).

O título da figura é "Distribuição das avaliações das compras em 2023" com tamanho 18 e sua cor CINZA_5.

Os rótulos do eixo x (xticks) precisa estar no range de 0 a 10 com bins de 11 bins e ter tamanho 12 com a cor AZUL_1. Os rótulos do eixo y (yticks) devem ter tamanho 12 com a cor AZUL_1 e os labels do eixo x e y (xlabel e ylabel) também devem ter tamanho 12 com a cor AZUL_1. Todas as bordas do gráfico devem ser suprimidas(['top','bottom','left','right]).

O grid do y tem que estar atrás do gráfico (ax.set_axisbelow(True)).

Vamos adicionar uma caixa de anotação (ax.annotate()) com o fundo na cor BRANCO e a borda na cor CINZA_3 dentro da área do plot. O texto da anotação é

f"Média de avaliações = {df['avaliacao_compra'].mean().round(2)}\n
Avaliação mais comum = {df['avaliacao_compra'].value_counts().idxmax()}\n
Frequência da mais comum = {df['avaliacao_compra'].value_counts().max()}"

Na cor AZUL_1. Adicione a caixa da anotação com o texto na posição xytext=(2, 1800) e coloque uma seta, na cor AZUL_1, saindo da caixa até a posição xy=(9, 2250). Utilize as bibliotecas de visualização de dados do Python para criar essa visualização.

No início do prompt, explicamos as cores e titulo que queremos para o gráfico. No eixo X, com as avaliações, traremos um range de 0 até 10, que são as avaliações possíveis, com o tamanho 12 e com o azul. No eixo do Y, em tamanho 12, traremos as frequências.

Os eixos X e Y terão títulos, já que estamos trabalhando com duas áreas numéricas, portanto é importante explicar. Além disso, solicitamos que o grid fique atrás.

Nisso, criaremos o nosso novo elemento, que é um ax.annotate( ), uma caixa de anotação em que colocaremos o texto que queremos apresentar no gráfico. Usaremos um fundo na cor branca e a borda na cor cinza para ser a caixa de texto.

Nesse texto teremos três informações interessantes, a média de avaliações, puxando do df['avaliacao_compra'] a média, arredondando as casas decimais. Além da avaliação mais comum através da leitura do .value_counts( ).

Quando criamos o .value_counts( ) das avaliação_compra, trazemos como índice os tipos de avaliação, de 0 até 10, e os valores vão ser a quantidade da frequência. Então, para conseguir puxar qual é a avaliação mais comum, precisamos usar o .idxmax( ), que irá conferir qual índice que tem o maior valor nas frequências. Já no caso da frequência mais comum, vamos só ler o ponto .idmax( ). Então, traremos isso também dentro do nosso texto.

Por fim, passaremos tudo isso na cor AZUL_1 e adicionaremos a caixa de anotação com o texto nessa posição xytext=(2. 1800). Adicionaremos uma seta que sairá da cor azul saindo da posição da caixa indo para a posição 9, até 2250.

Isso significa que estamos colocando a caixa em um ponto do gráfico mais ou menos situado à esquerda, puxando uma seta que vai da esquerda para a direita, um pouco mais para cima. Estamos adicionando a posição onde ficará o texto, e a posição onde terminará a seta.

Esse é o diferencial, não apenas a inserção do texto, mas onde vamos colocar a seta que está apontando a informação importante que estamos extraindo desse texto.

Ao enviar o prompt para o ChatGPT temos o seguinte código:

````python
import matplotlib.pyplot as plt
import seaborn as sns

def avaliacao_compra():
  # Configurações do gráfico
  plt.figure(figsize=(10, 6), facecolor=CINZA_1)
  ax = plt.axes()
  ax.set_facecolor(CINZA_1)

  # Plotar o histograma das avaliações
  sns.histplot(df['avaliacao_compra'], bins=11, color=VERMELHO_1, kde=False, ax=ax)

  # Adicionar rótulos e título
  plt.title('Distribuição das Avaliações das Compras em 2023', fontsize=18, color=CINZA_5)
  plt.xlabel('Avaliação', fontsize=12, color=AZUL_1)
  plt.ylabel('Frequência', fontsize=12, color=AZUL_1)
  plt.xticks(range(11), fontsize=12, color=AZUL_1)
  plt.yticks(fontsize=12, color=AZUL_1)
  ax.spines['top'].set_visible(False)
  ax.spines['right'].set_visible(False)
  ax.spines['left'].set_visible(False)
  ax.spines['bottom'].set_visible(False)
  ax.set_axisbelow(True)
  ax.yaxis.grid(color='gray', linestyle='--', linewidth=0.5)

  # Adicionar caixa de anotação
  media_avaliacao = df['avaliacao_compra'].mean().round(2)
  avaliacao_mais_comum = df['avaliacao_compra'].value_counts().idxmax()
  freq_avaliacao_mais_comum = df['avaliacao_compra'].value_counts().max()

  texto_anotacao = f"Média de avaliações = {media_avaliacao}\nAvaliação mais comum = {avaliacao_mais_comum}\nFrequência da mais comum = {freq_avaliacao_mais_comum}"

  ax.annotate(texto_anotacao, xy=(9, 2250), xytext=(2, 1800), fontsize=12, color=AZUL_1,
              bbox=dict(boxstyle="round,pad=0.3", edgecolor=CINZA_3, facecolor=BRANCO),
              arrowprops=dict(facecolor=AZUL_1, arrowstyle="->", connectionstyle="arc3,rad=0.3"))

  return plt
````

````python
avaliacao_compra()
````

image.png

Repare que conseguimos colocar uma caixa de texto, diferente do gráfico anterior. Nesse gráfico, adicionamos a informação e apontamos para o gráfico, na barra com maiores avaliações de compra.

Comparando ambos, conseguimos notar que o gráfico anterior não estava ruim. O que fizemos foram algumas melhorias para trazer a identidade da Zoop e a inserção das médias, avaliação e a frequência mais comum.

Conseguimos reduzir também alguns pontos que trazem um pouco mais de dificuldade na leitura e deixar esse gráfico mais sucinto, com a informação que precisamos deixar mais evidente para as partes interessadas.

Notaremos essa diferença visualmente.

Ao encaminhar o prompt para o ChatGPT, temos o seguinte código:

````python
import matplotlib.pyplot as plt
import seaborn as sns

def dist_idade_sexo():
    # Configurações do gráfico
    plt.figure(figsize=(10, 6), facecolor=CINZA_1)
    ax = plt.axes()
    ax.set_facecolor(CINZA_1)

    # Plotar o boxplot da distribuição da idade por sexo biológico
    sns.boxplot(data=df, x='sexo_biologico', y='idade', palette=[AZUL_1, AQUA_1], ax=ax)

    # Adicionar rótulos e título
    plt.title('Distribuição da Idade por Sexo Biológico dos(as) clientes', fontsize=18, color=CINZA_5)
    plt.xlabel('')
    plt.ylabel('Idade', fontsize=12, color=AZUL_1)
    plt.xticks(fontsize=12, color=AZUL_1)
    plt.yticks(fontsize=12, color=AZUL_1)
    ax.spines[['top', 'right', 'left', 'bottom']].set_visible(False)
    ax.set_axisbelow(True)
    ax.yaxis.grid(color='gray', linestyle='--', linewidth=0.5)

    # Adicionar caixas de anotação
    texto_anotacao1 = f"Mínimo = {df.groupby('sexo_biologico')['idade'].min()[1]}\n" \
                    f"1º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.25)[1]}\n" \
                    f"2º quartil = {df.groupby('sexo_biologico')['idade'].median()[1]}\n" \
                    f"3º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.75)[1]}\n" \
                    f"Máximo = {df.groupby('sexo_biologico')['idade'].max()[1]}"

    texto_anotacao2 = f"Mínimo = {df.groupby('sexo_biologico')['idade'].min()[0]}\n" \
                    f"1º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.25)[0]}\n" \
                    f"2º quartil = {df.groupby('sexo_biologico')['idade'].median()[0]}\n" \
                    f"3º quartil = {df.groupby('sexo_biologico')['idade'].quantile(.75)[0]}\n" \
                    f"Máximo = {df.groupby('sexo_biologico')['idade'].max()[0]}"

    ax.annotate(texto_anotacao1, xy=(-0.1, 45), xytext=(-0.4, 50), fontsize=12, color=AZUL_1,
                bbox=dict(boxstyle="round,pad=0.3", edgecolor=CINZA_3, facecolor=BRANCO),
                arrowprops=dict(facecolor=AZUL_1, arrowstyle="->", connectionstyle="arc3,rad=0.3"))

    ax.annotate(texto_anotacao2, xy=(0.9, 45), xytext=(0.6, 50), fontsize=12, color=AZUL_1,
                bbox=dict(boxstyle="round,pad=0.3", edgecolor=CINZA_3, facecolor=BRANCO),
                arrowprops=dict(facecolor=AZUL_1, arrowstyle="->", connectionstyle="arc3,rad=0.3"))

    return plt
````

Assim, conseguimos identificar que o valor mínimo e máximo no sexo masculino e feminino são os mesmos, que o intervalo está na faixa de 31 e 45 no sexo masculino e 32 e 45 no feminino.

Também conseguimos notar que um insight que podemos gerar é que entre 25 e 75% das pessoas usuárias que utilizam a Zoop estão no segmento dos millennials, ou seja, entre 31 e 45 anos. Dessa forma, sabemos um pouco mais sobre o público.

## Transformando os gráficos em imagens

Agora que geramos todos os gráficos solicitados pelas partes interessadas, vamos ler o último prompt que vai pegar cada um deles, transformar em uma função e trazê-los em formato de imagem do tipo PNG.

Agora, vamos para a nossa ultima etapa: salvar as visualizações como imagem. Coloquei cada um dos gráficos que geramos dentro de uma função e quero que seja aplicada a cada uma delas o seguinte procedimento:

nome_imagem = nome_funcao( )
nome_imagem.savefig('nome_imagem.png', bbox_inches='tight')
nome_imagem.close( )
As funções são:

metodos_pag( )
vendas_mensais( )
faturamento_por_categoria( )
vendas_tri_metodo )
cashback( )
avaliacao_compra( )
dist_idade_sexo( )
Troque a variável nome_imagem por grafico_ seguido do nome da função que gera o gráfico.

Utilize a linguagem Python para criar o código e não exiba as imagens.

O prompt diz para selecionar os gráficos e salvá-los como imagem. Nisso, é informado que estamos gerando uma função para fazer a aplicação desse processo. Isso significa pegar a função, passar para um savefig( ) para salvar a figura no formato PNG, deixando a imagem bem compactada. Por isso que usamos o bbox_inches.

Por fim fechamos a imagem, pois se formos salvar cada uma das figuras e não fechar a imagem, ele vai querer exibir na tela todas as imagens, e nós não queremos isso.

Passamos as funções responsáveis por cada um dos gráficos, sendo um total de sete, o metodo_pG( ), vendas_mensais( ), faturamento_por-categoria( ), vendas_tri_metodo( ), cashback( ), avaliação_compra( ) e dist_idade_sexo( ).

Pedimos para trocar esse nome da imagem por gráfico_ seguido do nome da função. Assim, temos um nome mais fácil de ser localizado na pasta que utilizaremos para gerar o relatório.

Ao enviar esse prompt no ChatGPT, temos o seguinte código:

````python
def salvar_grafico(grafico, nome):
    imagem = grafico()
    imagem.savefig(f'{nome}.png', bbox_inches='tight')
    imagem.close()

# Substitua as chamadas de função e nomes conforme necessário
salvar_grafico(metodos_pag, 'grafico_metodos_pagamento')
salvar_grafico(vendas_mensais, 'grafico_vendas_mensais')
salvar_grafico(faturamento_por_categoria, 'grafico_faturamento_categoria')
salvar_grafico(vendas_tri_metodo, 'grafico_vendas_trimestre_metodo')
salvar_grafico(cashback, 'grafico_cashback')
salvar_grafico(avaliacao_compra, 'grafico_avaliacao_compra')
salvar_grafico(dist_idade_sexo, 'grafico_distribuicao_idade_sexo')
````

Ele gera uma função chamada salvar_grafico( ) que recebe o tipo do gráfico e o nome. Assim, irá pegar o nome do gráfico, como metodos_pag, chamar a função e trazer para o objeto imagem, que vai salvar a figura com o nome da imagem, ajustar e fechar. Isso é feito em todos os gráficos e salva com esses nomes dentro da função salvar_grafico( ).

Ao rodar o código os gráficos não são exibidos. Isso acontece,, ao serem salvos na memória do Colab, para acessá-los, no menu lateral esquerdo, clicamos no último ícone chamado "Arquivos".

Feito isso, encontramos as sete figuras geradas. Ao clicá-las, conseguimos visualizar os gráficos. Para baixar, clicamos no ícone identificado por três pontos, ao lado da imagem, e depois em "Fazer Download".

Geramos a imagem e já podemos enviar para o time responsável por gerar o relatório. Mas que tal baixar essas imagens e utilizar uma IA, como o Gamma, para gerar um relatório utilizando os gráficos que criamos? Fica a dica!

## Temas

Ao criar uma apresentação no Gamma, você define um Tema para a sua apresentação, que pode ser personalizada.

image.png

Na atual versão do Gamma são seis opções:

1 - Plano de fundo do tema: aqui você escolhe como vai ser o fundo por trás dos seus cartões, que pode ser uma imagem, uma cor sólida, um gradiente, entre outras opções.

2 - Estilos de cartão: define a cor de fundo do próprio cartão. As cores do seu texto serão ajustadas automaticamente para terem contraste em relação ao fundo do cartão.

É possível decidir também o Estilo do cartão, podendo ser:

  - Padrão, com bordas arredondadas e sombras sutis;

  - Vidro, com transparência, tendo efeito de vidro fosco;

  - 3D delicado, com bordas arredondadas suaves com efeito 3D delicado;

  - Elegância, com design simples e minimalista com detalhes divertidos;

  - Sonhar acordado, com grandes bordas arredondadas, cores vibrantes e linhas grossas;

  - Quebra-gelo, com bordas limpas com cores monocromáticas profissionais.

3 - Cores de Destaque: Opção que apresenta primeiramente a cor de destaque principal, a qual pode ser somente uma cor, ou pode ser gradiente, com mais de uma cor, posteriormente temos as cores de destaque secundárias, que serão aplicadas a elementos decorativos do tema, como divisores de tabela, linhas do tempo, etc. E, se não adicionar nenhuma cor de destaque secundária, a principal ou o gradiente será utilizada. Ainda temos a legibilidade, sendo que por padrão o Gamma ajusta as cores para obedecerem às diretrizes de acessibilidade web, mas você pode desabilitar isso para utilizar as cores como estão.

4 - Imagens de destaque: As imagens de destaque são aquelas que podem ser associadas ao tema e usadas como decoração em seus cartões.

5 - Fontes: Aqui se determina as famílias tipográficas usadas em toda a apresentação. Pode-se configurar também o estilo e a cor das fontes dos títulos e do corpo.

6 - Logotipo: É possível arrastar o logotipo até a área de inserção ou clicar para fazer upload de seu logo.
